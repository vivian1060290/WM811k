```python
# =========================================================
# Two-Stage Wafer Classification (TRAIN + TEST) - PyTorch + DenseNet
# Swin-style overall architecture: samples -> unified Dataset -> generic train/eval -> repeated Greedy macro-F1
# ---------------------------------------------------------
# Stage1: Binary classification (None vs Defect)
# Stage2: Multiclass defect-only classification (8 classes)
# =========================================================




# =========================================================
# Package installation
# =========================================================
# !pip install numpy pandas matplotlib opencv-python pillow scikit-learn tqdm torch torchvision
# =========================================================




import os
import copy
import random
import zlib
from dataclasses import dataclass
from datetime import datetime
import cv2
import numpy as np
import pandas as pd
from PIL import Image
import torch
import torch.nn as nn
from torch.optim import Adam
from torch.optim.lr_scheduler import ReduceLROnPlateau
from torch.utils.data import Dataset, DataLoader, Subset
from torchvision import transforms, models
from sklearn.metrics import confusion_matrix, classification_report, f1_score
from tqdm import tqdm
# =========================================================
# 0) Config
# =========================================================
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)




torch.backends.cudnn.deterministic = False
torch.backends.cudnn.benchmark = torch.cuda.is_available()
if torch.cuda.is_available():
    torch.backends.cuda.matmul.allow_tf32 = True
    torch.backends.cudnn.allow_tf32 = True
    torch.set_float32_matmul_precision("high")




DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"[INFO] DEVICE = {DEVICE}")




# DenseNet 版本維持 32x32。這裡是「模型仍為 DenseNet」，但整體程式架構改成 Swin 程式碼那種統一流程。
IMAGE_SIZE = (64,64)
RESIZE_INTERPOLATION = transforms.InterpolationMode.NEAREST




BATCH_SIZE = 32
NUM_WORKERS = 0
print(f"[INFO] NUM_WORKERS = {NUM_WORKERS}")



DATASET_DIR = r"C:\Users\user\Desktop\MIR-WM811K\MIR-WM811K\Python\newdataset"
OUT_DIR =  r"C:\Users\user\Desktop\MIR-WM811K\MIR-WM811K\Python\densenet10"

TRAIN_DIR = os.path.join(DATASET_DIR, "train")
VAL_DIR = os.path.join(DATASET_DIR, "val")
TEST_DIR = os.path.join(DATASET_DIR, "test")




NONE_CLASS_NAME = "None"
CLASS_LIST = [
    "Center", "Donut", "Edge-loc", "Edge-ring", "Loc",
    "Random", "Scratch", "Near-Full", "None"
]
DEFECT_CLASS_LIST = [c for c in CLASS_LIST if c != NONE_CLASS_NAME]




STAGE1_WEIGHT_PATH = os.path.join(OUT_DIR, "stage1_densenet32_10.pt")
STAGE2_WEIGHT_PATH = os.path.join(OUT_DIR, "stage2_densenet32_10.pt")
OUT_TEST_CSV = os.path.join(OUT_DIR, "test_two_stage_predictions_densenet32_10.csv")
OUT_GREEDY_CSV = os.path.join(OUT_DIR, "greedy_defect_only_results_densenet32_10.csv")


# 正式訓練時，每個 epoch 額外輸出一次 checkpoint。
# Greedy trial 不會存 checkpoint，避免產生大量暫存模型。
STAGE1_EPOCH_CKPT_DIR = os.path.join(OUT_DIR, "stage1_epoch_checkpoints")
STAGE2_EPOCH_CKPT_DIR = os.path.join(OUT_DIR, "stage2_epoch_checkpoints")




EPOCHS_STAGE1 = 5
EPOCHS_STAGE2 = 15
THRESH_STAGE1 = 0.80
FORCE_RETRAIN_STAGE1 = False
FORCE_RETRAIN_STAGE2 = False




# Greedy augmentation search
RUN_GREEDY = True
USE_GREEDY_FOR_STAGE2 = True
# 與 Swin 版本對齊：Greedy trial 使用短訓練做候選 augmentation 比較。
TRIAL_EPOCHS = 2
CANDIDATES = ["dilation", "affine", "noise", "erase", "sharpen"]
MAX_ROUNDS = min(3, len(CANDIDATES))
MIN_GAIN = 0.003
GREEDY_REPEATS = 3


# 與修改後 Swin 版本對齊：Greedy 階段採用「每個 defect 類別最多抽 N 張」。
# 不足 N 張的類別會全部取用，避免普通隨機抽樣偏向資料量多的類別。
GREEDY_TRAIN_MAX_PER_CLASS = 1000
GREEDY_VAL_MAX_PER_CLASS = 500
GREEDY_FAST_EVAL = True




LR = 1e-4
WEIGHT_DECAY = 1e-4
DENSENET_DROP = 0.0


# =========================================================
# DenseNet 實驗主變因：loss function 中的 class weight strength
# =========================================================
# 注意：不要用 raw_w * 1.5 或 raw_w * 0.5 代表權重強度。
# 在 CrossEntropyLoss(weight=..., reduction="mean") 中，所有 class weight 同時乘上同一個倍率，
# 類別之間的相對比例不會改變，因此 Stage2 的實質意義很弱。
#
# 這裡改用 raw_w ** LOSS_WEIGHT_STRENGTH：
#   0.5：縮小類別權重差異，降低 class weight 補償強度
#   1.0：維持原始 class weight
#   1.5：放大類別權重差異，加強少數類別補償
#
# ===== 原本權重強度 =====
#LOSS_WEIGHT_STRENGTH = 1.0


# ===== 放大組：class weight strength = 1.5 =====
#LOSS_WEIGHT_STRENGTH = 1.5


# ===== 縮小組：class weight strength = 0.5 =====
LOSS_WEIGHT_STRENGTH = 1.0
# =========================================================
# 1) Utility
# =========================================================
def log(msg):
    now = datetime.now().strftime("%H:%M:%S")
    print(f"[{now}] {msg}", flush=True)








def set_all_seeds(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)









def round_float_columns(df, ndigits=2):
    """CSV 輸出前，將所有浮點數欄位統一四捨五入到小數點第二位。"""
    out = df.copy()
    float_cols = out.select_dtypes(include=["float", "float32", "float64"]).columns
    if len(float_cols) > 0:
        out[float_cols] = out[float_cols].round(ndigits)
    return out




def save_csv_round2(df, path):
    """安全輸出 CSV，並將浮點數欄位顯示到小數點第二位。"""
    os.makedirs(os.path.dirname(path), exist_ok=True)
    df_to_save = round_float_columns(df, ndigits=2)
    try:
        df_to_save.to_csv(path, index=False, encoding="utf-8-sig")
        print(f"[INFO] Saved CSV: {path}")
    except PermissionError:
        root, ext = os.path.splitext(path)
        alt_path = f"{root}_{datetime.now().strftime('%Y%m%d_%H%M%S')}{ext}"
        df_to_save.to_csv(alt_path, index=False, encoding="utf-8-sig")
        print(f"[WARN] Permission denied: {path}")
        print(f"[INFO] Saved CSV instead: {alt_path}")








def plot_cm(cm, labels, title, normalize="true", show_counts=False, save_path=None):
    try:
        import matplotlib.pyplot as plt
    except Exception as e:
        print(f"[WARN] Skip confusion-matrix plot: {title}. Reason: {e}")
        return




    cm = np.asarray(cm, dtype=np.int64)
    labels = list(labels)




    if normalize == "true":
        row_sum = cm.sum(axis=1, keepdims=True)
        cm_show = np.divide(
            cm.astype(np.float64),
            row_sum,
            out=np.zeros_like(cm, dtype=np.float64),
            where=row_sum != 0,
        ) * 100.0
        colorbar_label = "Row percentage (%)"
        vmin, vmax = 0, 100
    elif normalize == "none":
        cm_show = cm.astype(np.float64)
        colorbar_label = "Count"
        vmin, vmax = None, None
    else:
        raise ValueError("normalize must be 'true' or 'none'")




    fig_w = max(8, len(labels) * 0.9)
    fig_h = max(7, len(labels) * 0.75)
    plt.figure(figsize=(fig_w, fig_h))
    plt.imshow(cm_show, interpolation="nearest", cmap="Blues", vmin=vmin, vmax=vmax)
    plt.title(title)
    cbar = plt.colorbar()
    cbar.set_label(colorbar_label)
    plt.xticks(range(len(labels)), labels, rotation=45, ha="right")
    plt.yticks(range(len(labels)), labels)
    plt.xlabel("Predicted")
    plt.ylabel("True")




    for i in range(cm.shape[0]):
        row_total = cm[i].sum()
        for j in range(cm.shape[1]):
            count = int(cm[i, j])
            pct = (count / row_total * 100.0) if row_total > 0 else 0.0
            if normalize == "true":
                text = f"{pct:.1f}%"
                color = "white" if cm_show[i, j] > 50 else "black"
            else:
                text = str(count)
                color = "white" if cm_show[i, j] > cm_show.max() * 0.5 else "black"
            plt.text(j, i, text, ha="center", va="center", color=color, fontsize=8)




    plt.tight_layout()
    if save_path is not None:
        os.makedirs(os.path.dirname(save_path), exist_ok=True)
        plt.savefig(save_path, dpi=200, bbox_inches="tight")
        print(f"[INFO] Saved confusion matrix: {save_path}")
    plt.show()








def evaluate_stage1_binary_report(true_folder, stage1_pred_name, title="Stage1 None vs Defect"):
    y_true_s1 = np.where(np.asarray(true_folder).astype(str) == NONE_CLASS_NAME, "None", "Defect")
    y_pred_s1 = np.asarray(stage1_pred_name).astype(str)
    labels_s1 = ["None", "Defect"]
    print("\n==============================")
    print(title)
    print("==============================")
    print(classification_report(y_true_s1, y_pred_s1, labels=labels_s1, target_names=labels_s1, digits=2, zero_division=0))
    cm = confusion_matrix(y_true_s1, y_pred_s1, labels=labels_s1)
    plot_cm(cm, labels_s1, title, normalize="true", show_counts=False)
    return cm








def evaluate_final_report(true_folder, final_pred, title="Final Two-stage Confusion Matrix"):
    final_labels = [NONE_CLASS_NAME] + DEFECT_CLASS_LIST
    print("\n==============================")
    print("Final Two-stage Classification Report")
    print("==============================")
    print(classification_report(true_folder, final_pred, labels=final_labels, target_names=final_labels, digits=2, zero_division=0))
    final_macro_f1 = f1_score(true_folder, final_pred, labels=final_labels, average="macro", zero_division=0)
    print(f"Final macro-F1 = {final_macro_f1:.2f}")
    cm = confusion_matrix(true_folder, final_pred, labels=final_labels)
    plot_cm(cm, final_labels, title, normalize="true", show_counts=False)
    return final_macro_f1, cm








def stage1_false_negative_analysis(df_out, none_class_name="None"):
    print("\n==============================")
    print("Stage1 False-Negative Analysis (Defect -> None)")
    print("==============================")
    df_fn = df_out[
        (df_out["true_folder"].astype(str) != "") &
        (df_out["true_folder"].astype(str) != none_class_name) &
        (df_out["stage1_pred"].astype(str) == "None")
    ].copy()
    if len(df_fn) == 0:
        print("No defect samples were rejected as None by Stage1.")
        return
    print(f"Total Stage1 false negatives: {len(df_fn)}")
    print(df_fn["true_folder"].value_counts().sort_index())








def build_samples(root_dir, classes):
    samples = []
    class_to_idx = {c: i for i, c in enumerate(classes)}
    for cls in classes:
        cls_dir = os.path.join(root_dir, cls)
        if not os.path.isdir(cls_dir):
            print(f"[WARN] Class folder not found: {cls_dir}")
            continue
        for fn in sorted(os.listdir(cls_dir)):
            fp = os.path.join(cls_dir, fn)
            if not os.path.isfile(fp):
                continue
            samples.append({
                "path": os.path.normpath(fp),
                "relpath": os.path.join(cls, fn),
                "class_name": cls,
                "class_idx": class_to_idx[cls],
            })
    return samples, class_to_idx








def ensure_nonempty_samples(samples, name):
    if len(samples) == 0:
        raise RuntimeError(f"{name} has no samples. Please check dataset path and class folders.")








def stratified_cap_subset_samples(dataset, max_per_class, seed, save_csv_path=None):
    """Stratified per-class capped sampling for Greedy search.


    每個 class_name 最多抽 max_per_class 張。
    如果該類別資料數量不足 max_per_class，則全部使用。


    目的：
    1. 避免 Greedy trial 普通隨機抽樣偏向資料量多的 defect 類別。
    2. 讓每個 defect 類別都能參與 augmentation pipeline 評估。
    3. 控制 Greedy search 計算時間。
    """
    if max_per_class is None or max_per_class <= 0:
        return dataset


    class_to_indices = {}
    for idx, sample in enumerate(dataset.samples):
        class_name = sample["class_name"]
        class_to_indices.setdefault(class_name, []).append(idx)


    rng = np.random.default_rng(seed)
    selected_indices = []
    records = []


    print("\n==============================")
    print("Greedy stratified cap sampling")
    print("==============================")
    print(f"max_per_class = {max_per_class}")
    print(f"seed = {seed}")


    for class_name in sorted(class_to_indices.keys()):
        indices = np.asarray(class_to_indices[class_name], dtype=np.int64)
        rng.shuffle(indices)
        keep_n = min(int(max_per_class), len(indices))
        chosen = indices[:keep_n].tolist()
        selected_indices.extend(chosen)


        print(f"{class_name:<12} total={len(indices):5d} keep={keep_n:5d}")


        for original_idx in chosen:
            sample = dataset.samples[original_idx]
            records.append({
                "seed": seed,
                "class_name": class_name,
                "original_idx": original_idx,
                "path": sample["path"],
                "relpath": sample["relpath"],
            })


    selected_indices = sorted(selected_indices)
    print(f"Subset size = {len(selected_indices)} / {len(dataset)}")


    if save_csv_path is not None:
        os.makedirs(os.path.dirname(save_csv_path), exist_ok=True)
        save_csv_round2(pd.DataFrame(records), save_csv_path)


    return Subset(dataset, selected_indices)




def random_subset_samples(dataset, max_samples, seed):
    if max_samples is None or max_samples <= 0:
        return dataset
    n = len(dataset)
    max_items = min(int(max_samples), n)
    rng = np.random.default_rng(seed)
    indices = rng.choice(n, size=max_items, replace=False)
    return Subset(dataset, indices.tolist())








# =========================================================
# 2) Custom augmentation
# =========================================================

def wafer_affine(img, rng):
    """與 VGG 版本對齊的 affine augmentation。"""
    w, h = img.size
    scale = float(rng.uniform(0.90, 1.10))
    tx = float(rng.uniform(-0.05, 0.05) * w)
    ty = float(rng.uniform(-0.05, 0.05) * h)
    return img.transform(
        img.size,
        Image.AFFINE,
        (scale, 0.0, tx, 0.0, scale, ty),
        resample=Image.NEAREST,
    )








def wafer_dilation(img):
    img_np = np.array(img)
    kernel = np.ones((2, 2), np.uint8)
    return Image.fromarray(cv2.dilate(img_np, kernel))








def wafer_erosion(img):
    img_np = np.array(img)
    kernel = np.ones((2, 2), np.uint8)
    return Image.fromarray(cv2.erode(img_np, kernel))








def wafer_sharpen(img):
    img_np = np.array(img)
    kernel = np.array([[0, -1, 0], [-1, 5, -1], [0, -1, 0]])
    return Image.fromarray(cv2.filter2D(img_np, -1, kernel))








def wafer_noise(img, rng):
    img_np = np.array(img)
    noise = rng.normal(5, 30, img_np.shape).astype(np.int16)
    img_np = np.clip(img_np.astype(np.int16) + noise, 0, 255).astype(np.uint8)
    return Image.fromarray(img_np)








def wafer_random_erasing(img, rng, erase_width_ratio=0.95, erase_height_ratio=0.05, safe_center_ratio=0.4):
    img_np = np.array(img)
    h, w, _ = img_np.shape
    erase_w = max(1, min(int(w * erase_width_ratio), w))
    erase_h = max(1, min(int(h * erase_height_ratio), h))
    center_x, center_y = w // 2, h // 2
    safe_radius_x = int(w * safe_center_ratio / 2)
    safe_radius_y = int(h * safe_center_ratio / 2)
    x0, y0 = 0, 0
    for _ in range(50):
        x0 = int(rng.integers(0, max(1, w - erase_w + 1)))
        y0 = int(rng.integers(0, max(1, h - erase_h + 1)))
        erase_x1 = x0 + erase_w
        erase_y1 = y0 + erase_h
        safe_x0 = center_x - safe_radius_x
        safe_x1 = center_x + safe_radius_x
        safe_y0 = center_y - safe_radius_y
        safe_y1 = center_y + safe_radius_y
        overlap_safe = not (erase_x1 <= safe_x0 or x0 >= safe_x1 or erase_y1 <= safe_y0 or y0 >= safe_y1)
        if not overlap_safe:
            break
    img_np[y0:y0 + erase_h, x0:x0 + erase_w, :] = 0
    return Image.fromarray(img_np)








def aug_apply_wafer(img_pil, op, rng):
    if op == "dilation":
        return wafer_dilation(img_pil)
    if op == "affine":
        return wafer_affine(img_pil, rng)
    if op == "erosion":
        return wafer_erosion(img_pil)
    if op == "sharpen":
        return wafer_sharpen(img_pil)
    if op == "noise":
        return wafer_noise(img_pil, rng)
    if op == "erase":
        return wafer_random_erasing(img_pil, rng)
    return img_pil








# =========================================================
# 3) Dataset / DataLoader
# =========================================================
# DenseNet 從零訓練，不使用 ImageNet Normalize，保留第二份程式碼的 ToTensor 特性。
transform_densenet = transforms.Compose([
    transforms.Resize(IMAGE_SIZE, interpolation=RESIZE_INTERPOLATION),
    transforms.ToTensor(),
])








class WaferDataset(Dataset):
    def __init__(
        self,
        samples,
        transform,
        binary_none_name=None,
        defect_only=False,
        pipeline=None,
        apply_prob=0.7,
        seed=42,
    ):
        self.transform = transform
        self.binary_none_name = binary_none_name
        self.defect_only = defect_only
        self.pipeline = list(pipeline) if pipeline else []
        self.apply_prob = float(apply_prob)
        self.seed = int(seed)
        self.samples = [s for s in samples if (not defect_only or s["class_name"] != NONE_CLASS_NAME)]
        pipe_sig = "|".join(self.pipeline).encode("utf-8")
        self.pipe_hash = zlib.crc32(pipe_sig) & 0xFFFFFFFF
        self.defect_name_to_idx = {name: i for i, name in enumerate(DEFECT_CLASS_LIST)}




    def __len__(self):
        return len(self.samples)




    def _apply_pipeline(self, img, idx):
        if not self.pipeline:
            return img
        s = (self.seed * 1000003 + idx * 1009 + self.pipe_hash) & 0xFFFFFFFF
        rng = np.random.default_rng(s)
        out = img
        for op in self.pipeline:
            if rng.random() < self.apply_prob:
                out = aug_apply_wafer(out, op, rng)
        return out




    def __getitem__(self, idx):
        row = self.samples[idx]
        img = Image.open(row["path"]).convert("RGB")
        img = self._apply_pipeline(img, idx)
        x = self.transform(img)




        if self.binary_none_name is not None:
            y = 0 if row["class_name"] == self.binary_none_name else 1
        elif self.defect_only:
            y = self.defect_name_to_idx[row["class_name"]]
        else:
            y = row["class_idx"]




        meta = {
            "path": row["path"],
            "relpath": row["relpath"],
            "class_name": row["class_name"],
        }
        return x, y, meta








def build_loader(dataset, batch_size, shuffle=False, sampler=None):
    loader_kwargs = {}
    if NUM_WORKERS > 0:
        loader_kwargs["persistent_workers"] = True
        loader_kwargs["prefetch_factor"] = 2




    return DataLoader(
        dataset,
        batch_size=batch_size,
        shuffle=shuffle if sampler is None else False,
        sampler=sampler,
        num_workers=NUM_WORKERS,
        pin_memory=torch.cuda.is_available(),
        **loader_kwargs,
    )








# =========================================================
# 4) DenseNet model area
# =========================================================
class DenseNet32Classifier(nn.Module):
    def __init__(self, num_outputs, drop_rate=0.0):
        super().__init__()
        log(f"Creating DenseNet121-32x32 model, num_outputs={num_outputs}")
        self.backbone = models.densenet121(weights=None, drop_rate=drop_rate)
        self.backbone.features.conv0 = nn.Conv2d(3, 64, kernel_size=3, stride=1, padding=1, bias=False)
        self.backbone.features.pool0 = nn.Identity()
        in_features = self.backbone.classifier.in_features
        self.backbone.classifier = nn.Linear(in_features, num_outputs)
        log("DenseNet model created successfully.")




    def forward(self, x):
        return self.backbone(x)








def build_densenet_classifier(
    num_outputs,
    final_activation=None,
    loss_name=None,
    input_shape=(64,64, 3),
    drop=0.0,
    lr=1e-4,
):
    # 參數保留成類似第一份 build_swin_classifier 的介面，方便替換實驗。
    model = DenseNet32Classifier(num_outputs=num_outputs, drop_rate=drop).to(DEVICE)
    return model








def sanity_check_model_input():
    print("\n==============================")
    print("Sanity check: DenseNet input shape")
    print("==============================")
    tmp_model = build_densenet_classifier(num_outputs=2, drop=DENSENET_DROP)
    tmp_model.eval()
    with torch.no_grad():
        dummy = torch.randn(2, 3, IMAGE_SIZE[0], IMAGE_SIZE[1], device=DEVICE)
        out = tmp_model(dummy)
    print(f"Input shape  = {tuple(dummy.shape)}")
    print(f"Output shape = {tuple(out.shape)}")
    del tmp_model
    if torch.cuda.is_available():
        torch.cuda.empty_cache()








# =========================================================
# 5) Training flow
# =========================================================
@dataclass
class TrainOutput:
    best_state: dict
    best_val_loss: float








def safe_load_weights_if_compatible(model, weight_path):
    log(f"Checking weight file: {weight_path}")
    if not os.path.exists(weight_path):
        log(f"Weight file not found. Will train from scratch: {weight_path}")
        return False
    try:
        try:
            state = torch.load(weight_path, map_location=DEVICE, weights_only=True)
        except TypeError:
            state = torch.load(weight_path, map_location=DEVICE)
        model.load_state_dict(state)
        log(f"Loaded weights successfully: {weight_path}")
        return True
    except Exception as e:
        log(f"Weight load failed, will retrain: {weight_path}")
        log(f"Reason: {e}")
        return False








def save_weights(model, weight_path):
    os.makedirs(os.path.dirname(weight_path), exist_ok=True)
    torch.save(model.state_dict(), weight_path)
    log(f"Saved weights: {weight_path}")








def run_one_epoch(model, loader, criterion, optimizer=None, phase="train", epoch=None, epochs=None):
    is_train = optimizer is not None
    model.train(is_train)
    total_loss = 0.0
    total_correct = 0
    total_count = 0
    desc = f"{phase} epoch {epoch}/{epochs}" if epoch is not None and epochs is not None else phase
    pbar = tqdm(loader, desc=desc, leave=False, mininterval=5)




    for xb, yb, _ in pbar:
        xb = xb.to(DEVICE, non_blocking=True)
        yb = yb.to(DEVICE, non_blocking=True)




        if is_train:
            optimizer.zero_grad(set_to_none=True)




        with torch.set_grad_enabled(is_train):
            logits = model(xb)
            if logits.ndim == 1 or logits.shape[-1] == 1:
                logits_flat = logits.view(-1)
                y_float = yb.float()
                loss = criterion(logits_flat, y_float)
                probs = torch.sigmoid(logits_flat)
                preds = (probs >= 0.5).long()
            else:
                y_long = yb.long()
                loss = criterion(logits, y_long)
                preds = torch.argmax(logits, dim=1)




            if is_train:
                loss.backward()
                optimizer.step()




        total_loss += loss.item() * xb.size(0)
        total_correct += (preds == yb.long()).sum().item()
        total_count += xb.size(0)
        pbar.set_postfix({
            "loss": f"{total_loss / max(total_count, 1):.4f}",
            "acc": f"{total_correct / max(total_count, 1):.4f}",
            "seen": total_count,
        })




    return total_loss / max(total_count, 1), total_correct / max(total_count, 1)






def find_latest_checkpoint(checkpoint_dir, checkpoint_prefix):
    """
    找出 checkpoint_dir 裡面最新的 checkpoint。
    會依照檔名中的 epochXX 判斷最新 epoch。
    """
    if checkpoint_dir is None or not os.path.isdir(checkpoint_dir):
        return None

    candidates = []

    for fn in os.listdir(checkpoint_dir):
        if not fn.endswith(".pt"):
            continue

        if not fn.startswith(checkpoint_prefix):
            continue

        # 只接受完整 resume checkpoint
        if "_resume_epoch" not in fn:
            continue

        try:
            epoch_part = fn.split("_resume_epoch")[1].split("_")[0]
            ep = int(epoch_part)
            candidates.append((ep, os.path.join(checkpoint_dir, fn)))
        except Exception:
            continue

    if len(candidates) == 0:
        return None

    candidates.sort(key=lambda x: x[0])
    return candidates[-1][1]


def train_model_pt(
    model,
    train_loader,
    val_loader,
    criterion,
    lr,
    epochs,
    patience,
    save_each_epoch=False,
    checkpoint_dir=None,
    checkpoint_prefix="model",
    resume_from_checkpoint=False,
):
    optimizer = Adam(model.parameters(), lr=lr, weight_decay=WEIGHT_DECAY)

    scheduler = ReduceLROnPlateau(
        optimizer,
        mode="min",
        factor=0.5,
        patience=2,
        min_lr=1e-6
    )

    best_state = copy.deepcopy(model.state_dict())
    best_val_loss = np.inf
    wait = 0
    start_epoch = 1

    if save_each_epoch and checkpoint_dir is not None:
        os.makedirs(checkpoint_dir, exist_ok=True)

    # =====================================================
    # Resume training：從最新 checkpoint 接續訓練
    # =====================================================
    if resume_from_checkpoint and checkpoint_dir is not None:
        latest_ckpt = find_latest_checkpoint(checkpoint_dir, checkpoint_prefix)

        if latest_ckpt is not None:
            log(f"Resume checkpoint found: {latest_ckpt}")

            try:
                ckpt = torch.load(latest_ckpt, map_location=DEVICE)

                model.load_state_dict(ckpt["model_state"])
                optimizer.load_state_dict(ckpt["optimizer_state"])
                scheduler.load_state_dict(ckpt["scheduler_state"])

                best_state = ckpt["best_state"]
                best_val_loss = float(ckpt["best_val_loss"])
                wait = int(ckpt["wait"])

                start_epoch = int(ckpt["epoch"]) + 1

                log(
                    f"Resume training from epoch {start_epoch}/{epochs} | "
                    f"best_val_loss={best_val_loss:.4f} | wait={wait}"
                )

            except Exception as e:
                log("Resume checkpoint load failed. Start from epoch 1.")
                log(f"Reason: {e}")
                start_epoch = 1

        else:
            log("No resume checkpoint found. Start from epoch 1.")

    # 如果已經訓練完，不再重跑
    if start_epoch > epochs:
        log(f"Training already finished. start_epoch={start_epoch}, epochs={epochs}")
        model.load_state_dict(best_state)
        return TrainOutput(best_state=best_state, best_val_loss=float(best_val_loss))

    # =====================================================
    # Training loop
    # =====================================================
    for ep in range(start_epoch, epochs + 1):
        log(f"Start epoch {ep}/{epochs} training...")

        tr_loss, tr_acc = run_one_epoch(
            model,
            train_loader,
            criterion,
            optimizer=optimizer,
            phase="train",
            epoch=ep,
            epochs=epochs,
        )

        log(f"Start epoch {ep}/{epochs} validation...")

        va_loss, va_acc = run_one_epoch(
            model,
            val_loader,
            criterion,
            optimizer=None,
            phase="val",
            epoch=ep,
            epochs=epochs,
        )

        scheduler.step(va_loss)

        log(
            f"Epoch {ep:02d}/{epochs} | "
            f"train_loss={tr_loss:.4f} train_acc={tr_acc:.4f} | "
            f"val_loss={va_loss:.4f} val_acc={va_acc:.4f}"
        )

        # 更新最佳模型
        if va_loss < best_val_loss:
            best_val_loss = va_loss
            best_state = copy.deepcopy(model.state_dict())
            wait = 0
        else:
            wait += 1

        # =================================================
        # 每個 epoch 存完整 resume checkpoint
        # =================================================
        if save_each_epoch and checkpoint_dir is not None:
            resume_path = os.path.join(
                checkpoint_dir,
                f"{checkpoint_prefix}_resume_epoch{ep:02d}_valLoss{va_loss:.4f}.pt"
            )

            torch.save({
                "epoch": ep,
                "model_state": model.state_dict(),
                "optimizer_state": optimizer.state_dict(),
                "scheduler_state": scheduler.state_dict(),
                "best_state": best_state,
                "best_val_loss": best_val_loss,
                "wait": wait,
                "lr": lr,
                "epochs": epochs,
                "patience": patience,
            }, resume_path)

            log(f"Saved resume checkpoint: {resume_path}")

        # early stopping
        if wait >= patience:
            log("Early stopping triggered")
            break

    model.load_state_dict(best_state)
    return TrainOutput(best_state=best_state, best_val_loss=float(best_val_loss))






def predict_prob_stage1(model, loader):
    model.eval()
    probs = []
    with torch.no_grad():
        for xb, _, _ in tqdm(loader, desc="predict Stage1", leave=False, mininterval=5):
            xb = xb.to(DEVICE, non_blocking=True)
            logits = model(xb).view(-1)
            p = torch.sigmoid(logits)
            probs.append(p.detach().cpu().numpy())
    return np.concatenate(probs, axis=0)








def predict_stage2(model, loader):
    model.eval()
    probs = []
    metas = []
    with torch.no_grad():
        for xb, _, mb in tqdm(loader, desc="predict Stage2", leave=False, mininterval=5):
            xb = xb.to(DEVICE, non_blocking=True)
            logits = model(xb)
            p = torch.softmax(logits, dim=1)
            probs.append(p.detach().cpu().numpy())
            metas.extend([{k: mb[k][i] for k in mb} for i in range(len(mb["path"]))])
    return np.concatenate(probs, axis=0), metas








def collect_labels(loader):
    ys = []
    metas = []
    for _, yb, mb in tqdm(loader, desc="collect labels", leave=False, mininterval=5):
        ys.extend(yb.numpy().tolist())
        metas.extend([{k: mb[k][i] for k in mb} for i in range(len(mb["path"]))])
    return np.array(ys, dtype=np.int64), metas








def tune_stage1_threshold(model, val_loader, threshold_grid=None):
    if threshold_grid is None:
        threshold_grid = np.arange(0.10, 0.91, 0.05)
    val_prob = predict_prob_stage1(model, val_loader)
    y_val_true, _ = collect_labels(val_loader)
    print("\n==============================")
    print("Tune Stage1 Threshold on Validation Set")
    print("==============================")
    best_th = 0.50
    best_recall = -1.0
    best_f1 = -1.0
    for th in threshold_grid:
        y_pred = (val_prob >= th).astype(int)
        tp = np.sum((y_val_true == 1) & (y_pred == 1))
        fn = np.sum((y_val_true == 1) & (y_pred == 0))
        fp = np.sum((y_val_true == 0) & (y_pred == 1))
        recall_defect = tp / (tp + fn + 1e-8)
        precision_defect = tp / (tp + fp + 1e-8)
        f1_defect = 2 * precision_defect * recall_defect / (precision_defect + recall_defect + 1e-8)
        print(f"th={th:.2f} | recall={recall_defect:.2f} | precision={precision_defect:.2f} | f1={f1_defect:.2f}")
        if (recall_defect > best_recall) or (np.isclose(recall_defect, best_recall) and f1_defect > best_f1):
            best_recall = recall_defect
            best_f1 = f1_defect
            best_th = float(th)
    print(f"\n[Selected Stage1 threshold] {best_th:.2f}")
    return best_th








def evaluate_stage2(model, loader, title):
    probs, metas = predict_stage2(model, loader)
    y_pred = np.argmax(probs, axis=1)
    y_true, _ = collect_labels(loader)
    print("\n==============================")
    print(title)
    print("==============================")
    print(classification_report(y_true, y_pred, labels=list(range(len(DEFECT_CLASS_LIST))), target_names=DEFECT_CLASS_LIST, digits=2, zero_division=0))
    macro_f1 = f1_score(y_true, y_pred, average="macro", zero_division=0)
    print(f"{title} macro-F1 = {macro_f1:.2f}")
    cm = confusion_matrix(y_true, y_pred, labels=list(range(len(DEFECT_CLASS_LIST))))
    plot_cm(cm, DEFECT_CLASS_LIST, title, normalize="true", show_counts=False)
    return macro_f1








def evaluate_stage2_macro_f1_only(model, loader):
    model.eval()
    all_pred = []
    all_true = []
    with torch.no_grad():
        for xb, yb, _ in loader:
            xb = xb.to(DEVICE, non_blocking=True)
            logits = model(xb)
            pred = torch.argmax(logits, dim=1).detach().cpu().numpy()
            all_pred.extend(pred.tolist())
            all_true.extend(yb.numpy().tolist())
    return f1_score(np.asarray(all_true, dtype=np.int64), np.asarray(all_pred, dtype=np.int64), labels=list(range(len(DEFECT_CLASS_LIST))), average="macro", zero_division=0)








# =========================================================
# 6) Greedy augmentation search for Stage2
# =========================================================
def evaluate_stage2_pipeline_once(train_samples, val_samples, pipeline, trial_epochs, repeat_seed):
    set_all_seeds(repeat_seed)




    trial_train_ds = WaferDataset(
        train_samples,
        transform_densenet,
        defect_only=True,
        pipeline=pipeline,
        apply_prob=0.7,
        seed=repeat_seed,
    )
    trial_train_ds = stratified_cap_subset_samples(
        trial_train_ds,
        max_per_class=GREEDY_TRAIN_MAX_PER_CLASS,
        seed=repeat_seed,
        save_csv_path=os.path.join(OUT_DIR, f"greedy_train_subset_seed_{repeat_seed}.csv"),
    )




    trial_val_ds = WaferDataset(
        val_samples,
        transform_densenet,
        defect_only=True,
        pipeline=None,
        seed=SEED,
    )
    trial_val_ds = stratified_cap_subset_samples(
        trial_val_ds,
        max_per_class=GREEDY_VAL_MAX_PER_CLASS,
        seed=SEED,
        save_csv_path=os.path.join(OUT_DIR, f"greedy_val_subset_seed_{SEED}.csv"),
    )




    trial_train_loader = build_loader(trial_train_ds, BATCH_SIZE, shuffle=True)
    trial_val_loader = build_loader(trial_val_ds, BATCH_SIZE, shuffle=False)




    trial_model = build_densenet_classifier(num_outputs=len(DEFECT_CLASS_LIST), final_activation=None, loss_name="ce", drop=DENSENET_DROP)
    criterion = nn.CrossEntropyLoss(weight=weight_tensor_s2)




    train_model_pt(
        trial_model,
        trial_train_loader,
        trial_val_loader,
        criterion,
        lr=LR,
        epochs=trial_epochs,
        patience=max(1, trial_epochs),
    )




    if GREEDY_FAST_EVAL:
        score = evaluate_stage2_macro_f1_only(trial_model, trial_val_loader)
        print(f"[Greedy trial] pipeline={pipeline}, seed={repeat_seed}, macro-F1={score:.2f}")
    else:
        score = evaluate_stage2(trial_model, trial_val_loader, title=f"Greedy trial pipeline={pipeline}, seed={repeat_seed}")




    del trial_model
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
    return float(score)








def evaluate_stage2_pipeline_repeated(train_samples, val_samples, pipeline, trial_epochs):
    scores = []
    for repeat_id in range(GREEDY_REPEATS):
        repeat_seed = SEED + 1000 * repeat_id
        score = evaluate_stage2_pipeline_once(train_samples, val_samples, pipeline, trial_epochs, repeat_seed=repeat_seed)
        scores.append(score)
    scores = np.asarray(scores, dtype=np.float64)
    mean_score = float(scores.mean())
    std_score = float(scores.std(ddof=1)) if len(scores) > 1 else 0.0
    print(f"[Greedy repeated] pipeline={pipeline}, scores={np.round(scores, 2).tolist()}, mean={mean_score:.2f}, std={std_score:.2f}")
    return mean_score, std_score, scores.tolist()






def greedy_search_stage2(train_samples, val_samples):
    """
    VGG 版本同款 Greedy augmentation search。

    流程：
        1. 先評估 baseline，也就是不使用資料增強 pipeline=[]。
        2. 每一輪嘗試 selected + [candidate]。
        3. 從當輪候選中選出 mean macro-F1 最高者。
        4. 若該候選相較目前 best_score 的提升 >= MIN_GAIN，則保留。
        5. 否則停止搜尋。
        6. 最後輸出 selected pipeline 與 greedy CSV。
    """
    print("\n==============================")
    print("Greedy Stage2 Augmentation Selection")
    print("VGG-style baseline + round-wise candidate selection")
    print("==============================")

    rows = []
    selected = []
    remaining = list(CANDIDATES)

    base_mean, base_std, base_scores = evaluate_stage2_pipeline_repeated(
        train_samples,
        val_samples,
        [],
        TRIAL_EPOCHS,
    )
    best_score = base_mean
    best_std = base_std

    rows.append({
        "round": 0,
        "candidate": "(baseline)",
        "candidate_op": "(baseline)",
        "pipeline": "",
        "n_augmentations": 0,
        "mean_macro_f1": round(float(base_mean), 2),
        "std_macro_f1": round(float(base_std), 2),
        "scores": "|".join(f"{float(s):.2f}" for s in base_scores),
        "gain": 0.0,
        "selected": 0,
        "decision": "baseline",
    })

    print(f"[Baseline] macro-F1={base_mean:.2f}±{base_std:.2f}")

    for rnd in range(1, MAX_ROUNDS + 1):
        print(f"\n===== Round {rnd}/{MAX_ROUNDS} =====")
        results = []

        for cand in remaining:
            pipe = selected + [cand]

            mean_score, std_score, scores = evaluate_stage2_pipeline_repeated(
                train_samples,
                val_samples,
                pipe,
                TRIAL_EPOCHS,
            )

            gain = mean_score - best_score

            print(
                f"{pipe} | macro-F1={mean_score:.2f}±{std_score:.2f} "
                f"gain={gain:+.2f}"
            )

            rows.append({
                "round": rnd,
                "candidate": cand,
                "candidate_op": cand,
                "pipeline": "|".join(pipe),
                "n_augmentations": len(pipe),
                "mean_macro_f1": round(float(mean_score), 2),
                "std_macro_f1": round(float(std_score), 2),
                "scores": "|".join(f"{float(s):.2f}" for s in scores),
                "gain": round(float(gain), 2),
                "selected": 0,
                "decision": "candidate",
            })
            results.append((cand, mean_score, std_score))

        if len(results) == 0:
            print("[STOP] no remaining augmentation candidates.")
            break

        results.sort(key=lambda x: (-x[1], x[2]))
        best_cand, cand_score, cand_std = results[0]
        mean_gain = cand_score - best_score

        print(f"[Round {rnd}] best_candidate={best_cand}, gain={mean_gain:+.2f}")

        if mean_gain < MIN_GAIN:
            print(f"[STOP] best gain {mean_gain:+.2f} < MIN_GAIN={MIN_GAIN}")
            for row in reversed(rows):
                if row["round"] == rnd and row["candidate_op"] == best_cand:
                    row["decision"] = "stop_not_enough_gain"
                    break
            break

        selected.append(best_cand)
        remaining.remove(best_cand)
        best_score = cand_score
        best_std = cand_std

        print(
            f"[SELECT] add {best_cand} -> "
            f"selected={selected}, best_macro-F1={best_score:.2f}±{best_std:.2f}"
        )

        for row in reversed(rows):
            if row["round"] == rnd and row["candidate_op"] == best_cand:
                row["selected"] = 1
                row["decision"] = "selected"
                break

        if len(remaining) == 0:
            print("[STOP] no remaining augmentation candidates.")
            break

    df_greedy = pd.DataFrame(rows)
    save_csv_round2(df_greedy, OUT_GREEDY_CSV)

    print(f"[INFO] Saved greedy results: {OUT_GREEDY_CSV}")
    print(f"[INFO] Final selected pipeline: {selected}")

    return selected








# =========================================================
# 7) Main pipeline
# =========================================================
os.makedirs(OUT_DIR, exist_ok=True)
log("Build sample lists.")
train_samples, train_class_to_idx = build_samples(TRAIN_DIR, CLASS_LIST)
val_samples, val_class_to_idx = build_samples(VAL_DIR, CLASS_LIST)
test_samples, test_class_to_idx = build_samples(TEST_DIR, CLASS_LIST)




ensure_nonempty_samples(train_samples, "train_samples")
ensure_nonempty_samples(val_samples, "val_samples")
ensure_nonempty_samples(test_samples, "test_samples")




print(f"[INFO] train samples = {len(train_samples)}")
print(f"[INFO] val samples   = {len(val_samples)}")
print(f"[INFO] test samples  = {len(test_samples)}")
print(f"[INFO] train_class_to_idx = {train_class_to_idx}")




# Build base datasets/loaders.
log("Build datasets and dataloaders.")
train_ds_bin = WaferDataset(train_samples, transform_densenet, binary_none_name=NONE_CLASS_NAME)
val_ds_bin = WaferDataset(val_samples, transform_densenet, binary_none_name=NONE_CLASS_NAME)
test_ds_bin = WaferDataset(test_samples, transform_densenet, binary_none_name=NONE_CLASS_NAME)




train_ds_def = WaferDataset(train_samples, transform_densenet, defect_only=True)
val_ds_def = WaferDataset(val_samples, transform_densenet, defect_only=True)
test_ds_all = WaferDataset(test_samples, transform_densenet, defect_only=False)




train_bin_flow = build_loader(train_ds_bin, BATCH_SIZE, shuffle=True)
val_bin_flow = build_loader(val_ds_bin, BATCH_SIZE, shuffle=False)
test_bin_flow = build_loader(test_ds_bin, BATCH_SIZE, shuffle=False)
val_def_flow = build_loader(val_ds_def, BATCH_SIZE, shuffle=False)
test_all_flow = build_loader(test_ds_all, BATCH_SIZE, shuffle=False)




# Compute class weights.
# =========================================================
# 這一段保留第二份 DenseNet 程式碼的「改變 loss function」設計。
# 注意：第一份 Swin 實驗主要改 learning rate；這份 DenseNet 實驗不要改成那種設定。
# DenseNet 的實驗重點維持在 BCEWithLogitsLoss / CrossEntropyLoss 的權重倍率調整。
# =========================================================
log("Compute Stage1 and Stage2 class weights.")




print("\n==============================")
print("Compute class weight for Stage1 (None vs Defect)")
print("==============================")
stage1_counts = {0: 0, 1: 0}
for s in train_samples:
    y = 0 if s["class_name"] == NONE_CLASS_NAME else 1
    stage1_counts[y] += 1




n_none = stage1_counts[0]
n_def = stage1_counts[1]




print(f"Stage1 None count   = {n_none}")
print(f"Stage1 Defect count = {n_def}")




base_pos_weight = n_none / max(n_def, 1)




# =========================================================
# Stage1 loss weight strength
# =========================================================
# base_pos_weight 是原始 Defect 類別權重。
# 這裡使用 ** LOSS_WEIGHT_STRENGTH，而不是整體乘 1.5 / 0.5。
#
# 可在 Config 區切換：
#   LOSS_WEIGHT_STRENGTH = 1.0  # 原始權重
#   LOSS_WEIGHT_STRENGTH = 1.5  # 放大權重差異
#   LOSS_WEIGHT_STRENGTH = 0.5  # 縮小權重差異
pos_weight_stage1 = torch.tensor(
    [base_pos_weight ** LOSS_WEIGHT_STRENGTH],
    dtype=torch.float32,
).to(DEVICE)




print("Stage1 base_pos_weight =", base_pos_weight)
print("Stage1 LOSS_WEIGHT_STRENGTH =", LOSS_WEIGHT_STRENGTH)
print("Stage1 pos_weight =", pos_weight_stage1.item())




print("\n==============================")
print("Compute class weights for Stage2 (defect-only)")
print("==============================")
counts_s2 = np.zeros(len(DEFECT_CLASS_LIST), dtype=np.float64)
defect_to_idx = {name: i for i, name in enumerate(DEFECT_CLASS_LIST)}
for s in train_samples:
    if s["class_name"] != NONE_CLASS_NAME:
        counts_s2[defect_to_idx[s["class_name"]]] += 1




n_samples_s2 = counts_s2.sum()
K_s2 = len(DEFECT_CLASS_LIST)
counts_safe = np.where(counts_s2 == 0, 1.0, counts_s2)
raw_w = n_samples_s2 / (K_s2 * counts_safe)




# =========================================================
# Stage2 loss weight strength
# =========================================================
# raw_w 是根據各 defect 類別數量自動計算的原始 class weight。
# 這裡使用 raw_w ** LOSS_WEIGHT_STRENGTH 來真正改變類別權重差異：
#   strength = 0.5：縮小類別權重差距
#   strength = 1.0：維持原始類別權重
#   strength = 1.5：放大類別權重差距
#
# 不建議使用 raw_w * 1.5 或 raw_w * 0.5，因為全部類別同時乘同一數值時，
# 各類別之間的相對比例不變，對 CrossEntropyLoss 的 Stage2 實質影響很弱。
class_weights_s2_np = raw_w ** LOSS_WEIGHT_STRENGTH
class_weights_s2 = torch.tensor(class_weights_s2_np, dtype=torch.float32).to(DEVICE)




weight_tensor_s2 = class_weights_s2
print("-" * 70)
print(f"Stage2 LOSS_WEIGHT_STRENGTH = {LOSS_WEIGHT_STRENGTH}")
for idx, name in enumerate(DEFECT_CLASS_LIST):
    print(
        f"{name:<12} count={int(counts_s2[idx]):5d}  "
        f"raw_w={raw_w[idx]:8.4f}  final_w={class_weights_s2[idx].item():8.4f}"
    )
print("-" * 70)




# Optional sanity check.
sanity_check_model_input()




# Stage1 train/load.
log("Stage1: None vs Defect.")
stage1 = build_densenet_classifier(num_outputs=1, final_activation=None, loss_name="bce", drop=DENSENET_DROP)
criterion_stage1 = nn.BCEWithLogitsLoss(pos_weight=pos_weight_stage1)
loaded_stage1 = False
if not FORCE_RETRAIN_STAGE1:
    loaded_stage1 = safe_load_weights_if_compatible(stage1, STAGE1_WEIGHT_PATH)
if not loaded_stage1:
    log("Stage1 weights not loaded. Start Stage1 training.")
    train_model_pt(
        stage1,
        train_bin_flow,
        val_bin_flow,
        criterion_stage1,
        lr=LR,
        epochs=EPOCHS_STAGE1,
        patience=4,
        save_each_epoch=True,
        checkpoint_dir=STAGE1_EPOCH_CKPT_DIR,
        checkpoint_prefix=f"stage1_densenet_strength{LOSS_WEIGHT_STRENGTH}",
        resume_from_checkpoint=True,

    )
    save_weights(stage1, STAGE1_WEIGHT_PATH)
else:
    log("Stage1 training skipped because saved weights were loaded.")




# Stage1 threshold tuning.
log("Tune Stage1 threshold on validation set.")
THRESH_STAGE1 = tune_stage1_threshold(stage1, val_bin_flow)
print(f"[INFO] THRESH_STAGE1 = {THRESH_STAGE1:.2f}")




# Greedy augmentation search for Stage2.
log("Greedy augmentation search for Stage2.")
greedy_selected = []
if RUN_GREEDY:
    greedy_selected = greedy_search_stage2(train_samples, val_samples)
else:
    log("Greedy augmentation search skipped because RUN_GREEDY=False.")




stage2_pipeline = greedy_selected if USE_GREEDY_FOR_STAGE2 else []
print(f"[INFO] Stage2 training pipeline = {stage2_pipeline}")




# Stage2 train loader with selected pipeline.
log("Build Stage2 train loader with selected augmentation pipeline.")
train_ds_stage2 = WaferDataset(
    train_samples,
    transform_densenet,
    defect_only=True,
    pipeline=stage2_pipeline,
    apply_prob=0.7,
    seed=SEED,
)
train_def_flow = build_loader(train_ds_stage2, BATCH_SIZE, shuffle=True)
print(f"[INFO] Stage2 train dataset size = {len(train_ds_stage2)}")
print(f"[INFO] Stage2 pipeline = {stage2_pipeline}")




# Stage2 train/load.
log("Stage2: Defect-only 8-class classification.")
stage2 = build_densenet_classifier(num_outputs=len(DEFECT_CLASS_LIST), final_activation=None, loss_name="ce", drop=DENSENET_DROP)
criterion_stage2 = nn.CrossEntropyLoss(weight=weight_tensor_s2)
loaded_stage2 = False
if not FORCE_RETRAIN_STAGE2:
    loaded_stage2 = safe_load_weights_if_compatible(stage2, STAGE2_WEIGHT_PATH)
if not loaded_stage2:
    log("Stage2 weights not loaded. Start Stage2 training.")
    train_model_pt(
        stage2,
        train_def_flow,
        val_def_flow,
        criterion_stage2,
        lr=LR,
        epochs=EPOCHS_STAGE2,
        patience=5,
        save_each_epoch=True,
        checkpoint_dir=STAGE2_EPOCH_CKPT_DIR,
        checkpoint_prefix=f"stage2_densenet_strength{LOSS_WEIGHT_STRENGTH}",
        resume_from_checkpoint=True,
    )
    save_weights(stage2, STAGE2_WEIGHT_PATH)
else:
    log("Stage2 training skipped because saved weights were loaded.")




# Stage2 validation evaluation.
log("Stage2 validation evaluation.")
stage2_val_macro_f1 = evaluate_stage2(stage2, val_def_flow, title="Stage2 Validation Result")
print(f"[INFO] Stage2 validation macro-F1 = {stage2_val_macro_f1:.2f}")




# Two-stage test prediction.
log("Two-stage test prediction.")
print("\n==============================")
print("Two-stage Test")
print("==============================")




stage1_test_prob = predict_prob_stage1(stage1, test_bin_flow)
stage1_pred_bin = (stage1_test_prob >= THRESH_STAGE1).astype(int)
stage1_pred_name = np.where(stage1_pred_bin == 1, "Defect", "None")




stage2_test_prob, metas = predict_stage2(stage2, test_all_flow)
stage2_pred_idx = np.argmax(stage2_test_prob, axis=1)
stage2_pred_name = np.array([DEFECT_CLASS_LIST[i] for i in stage2_pred_idx])
stage2_conf = np.max(stage2_test_prob, axis=1)




final_pred = np.where(stage1_pred_bin == 0, NONE_CLASS_NAME, stage2_pred_name)
true_folder = np.array([m["class_name"] for m in metas])




df_out = pd.DataFrame({
    "path": [m["path"] for m in metas],
    "relpath": [m["relpath"] for m in metas],
    "file": [os.path.basename(m["path"]) for m in metas],
    "true_folder": true_folder,
    "stage1_prob_defect": stage1_test_prob,
    "stage1_pred": stage1_pred_name,
    "stage2_pred": stage2_pred_name,
    "stage2_conf": stage2_conf,
    "final_pred": final_pred,
})




save_csv_round2(df_out, OUT_TEST_CSV)




# Reports.
log("Stage1 binary test report and confusion matrix.")
evaluate_stage1_binary_report(true_folder, stage1_pred_name, title="TEST Stage1 None vs Defect")




log("Final two-stage report and confusion matrix.")
evaluate_final_report(true_folder, final_pred, title="TEST Final Two-stage Confusion Matrix")




log("Stage1 false-negative analysis.")
stage1_false_negative_analysis(df_out, none_class_name=NONE_CLASS_NAME)




print("\n[INFO] Done.")
```
