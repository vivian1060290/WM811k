```python
# =========================================================
# Three Models Two-Stage Ensemble
# Greedy-Based Selection Algorithm Version
# ---------------------------------------------------------
# Stage1: None / Defect binary classification
# Stage2: 8-class defect-only classification
# Models: Official CNN VGG11-BN-64 / DenseNet / Swin Transformer
# =========================================================

import os
import copy
import random
import zlib
from dataclasses import dataclass, field
from datetime import datetime
from itertools import combinations
from typing import Dict, List, Optional, Tuple

import cv2
import numpy as np
import pandas as pd
from PIL import Image

import torch
import torch.nn as nn
from torch.optim import Adam
from torch.optim.lr_scheduler import ReduceLROnPlateau
from torch.utils.data import Dataset, DataLoader

from torchvision import transforms, models

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    f1_score,
    precision_recall_fscore_support,
)

from tqdm import tqdm

try:
    import timm
except Exception:
    timm = None





# =========================================================
# 1. Global Config：資料路徑、類別、執行模式
# =========================================================
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)


torch.backends.cudnn.deterministic = False
torch.backends.cudnn.benchmark = torch.cuda.is_available()
if torch.cuda.is_available():
    torch.backends.cuda.matmul.allow_tf32 = True
    torch.backends.cudnn.allow_tf32 = True
    torch.set_float32_matmul_precision("high")


DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"[INFO] DEVICE = {DEVICE}")


# ===== 你的資料集根目錄 =====
# 需要包含 train / val / test 三個資料夾，每個資料夾內再依類別分資料夾。
DATASET_DIR = r"C:\Users\user\Desktop\MIR-WM811K\MIR-WM811K\Python\newdataset"
TRAIN_DIR = os.path.join(DATASET_DIR, "train")
VAL_DIR = os.path.join(DATASET_DIR, "val")
TEST_DIR = os.path.join(DATASET_DIR, "test")


# ===== 整合實驗輸出資料夾 =====
ENSEMBLE_OUT_DIR = r"C:\Users\user\Desktop\MIR-WM811K\MIR-WM811K\Python\ensemble"
os.makedirs(ENSEMBLE_OUT_DIR, exist_ok=True)


NONE_CLASS_NAME = "None"
CLASS_LIST = [
    "Center", "Donut", "Edge-loc", "Edge-ring", "Loc",
    "Random", "Scratch", "Near-Full", "None"
]
DEFECT_CLASS_LIST = [c for c in CLASS_LIST if c != NONE_CLASS_NAME]


# ===== 訓練 / 載入模式 =====
# False：嚴格使用既有 .pt 權重檔，缺權重就報錯。
# True ：如果找不到 .pt，會使用該模型指定的最佳資料增強 pipeline 重新訓練。
TRAIN_SINGLE_MODEL_IF_MISSING = True
FORCE_RETRAIN_SINGLE_MODELS = False


# ===== 正式訓練 epochs =====
EPOCHS_STAGE1 = 5
EPOCHS_STAGE2 = 15
BATCH_SIZE = 32
NUM_WORKERS = 0
WEIGHT_DECAY = 1e-4
APPLY_AUG_PROB = 0.7

# ===== 輸出與 Greedy 比較小數位設定 =====
# 所有 CSV/JSON 中的浮點指標統一輸出到小數點第二位。
# Greedy 是否加入模型時也以小數點第二位後的 F1 ratio 比較。
ROUND_OUTPUT_DECIMALS = 2


# ===== Stage1 threshold 搜尋 =====
THRESH_GRID = np.arange(0.10, 0.91, 0.05)


# ===== Greedy model search =====
# 這裡是「模型組合」搜尋，不是重新做 augmentation 搜尋。
# 單一模型的 augmentation 已經由 BEST_STAGE1_PIPELINE / BEST_STAGE2_PIPELINE 指定。
GREEDY_STAGE1_MODELS_CSV = os.path.join(ENSEMBLE_OUT_DIR, "greedy_stage1_model_combination.csv")
GREEDY_STAGE2_MODELS_CSV = os.path.join(ENSEMBLE_OUT_DIR, "greedy_stage2_model_combination.csv")
OUT_TEST_CSV = os.path.join(ENSEMBLE_OUT_DIR, "test_two_stage_cnn_densenet_swin.csv")
OUT_SAMPLE60_CSV = os.path.join(ENSEMBLE_OUT_DIR, "sample60_cnn_densenet_swin.csv")
OUT_FINAL_CONFIG_JSON = os.path.join(ENSEMBLE_OUT_DIR, "final_cnn_densenet_swin_config.json")

# ===== 弱勢類別分析 =====
# 從目前最佳單一模型表現較差的類別中挑選前 K 個，
# 比較 Greedy-Based Selection 選出的模型集合是否優於單一模型。
WEAK_CLASS_TOP_K = 3
OUT_WEAK_CLASS_CSV = os.path.join(ENSEMBLE_OUT_DIR, "weak_class_greedy_vs_single_comparison.csv")


# ===== sample60 =====
SAMPLE60_DIR = os.path.join(ENSEMBLE_OUT_DIR, "sample60_fixed")
FORCE_REBUILD_SAMPLE60 = False


RESIZE_INTERPOLATION = transforms.InterpolationMode.NEAREST
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD = [0.229, 0.224, 0.225]




# =========================================================
# 2. Best Model Config：三個模型集中在這裡修改
# =========================================================
# 你之後主要只需要改這區：
#   1) OUT_DIR
#   2) stage1_weight_path / stage2_weight_path
#   3) best_stage1_pipeline / best_stage2_pipeline
#   4) 模型超參數，例如 DenseNet LOSS_WEIGHT_STRENGTH、CNN dropout
#
# 注意：目前 best_stage*_pipeline 先放空 list。
#      請把各單一模型 Greedy Search 輸出的最佳增強組合填進來。
#      例如：['sharpen', 'noise']、['dilation']、['erase']。
# =========================================================


DENSENET_OUT_DIR = r"C:\Users\user\Desktop\MIR-WM811K\MIR-WM811K\Python\densenet05"
SWIN_OUT_DIR = r"C:\Users\user\Desktop\MIR-WM811K\MIR-WM811K\Python\swinlr5"
CNN_OUT_DIR = r"C:\Users\user\Desktop\MIR-WM811K\MIR-WM811K\Python\official_cnn_vgg11bn_64"


# ===== DenseNet 最佳設定 =====
# 依你目前 densenet05 程式：LOSS_WEIGHT_STRENGTH = 0.5，LR = 1e-4，input 64x64。
DENSENET_LOSS_WEIGHT_STRENGTH = 0.5


# ===== Swin 最佳設定 =====
# 依你之前 Swin 版本：timm swin_tiny_patch4_window7_224，將 64x64 原圖 resize 成 224x224。
USE_IMAGENET_PRETRAINED_SWIN = True
SWIN_MODEL_NAME = "swin_tiny_patch4_window7_224"
SWIN_DROP = 0.1


# ===== Official CNN VGG11-BN-64 最佳設定 =====
# 使用 torchvision 官方 VGG11-BN 架構，輸入尺寸為 64x64。
# 這是純 CNN 架構，不使用 ResNet residual connection，也不是 Transformer / Mamba。
# 權重來源為 official_cnn_vgg11bn_64 單一模型程式訓練出的 .pt。
USE_IMAGENET_PRETRAINED_CNN = False
CNN_IMAGE_SIZE = (64, 64)
CNN_LR = 1e-4
CNN_DROPOUT = 0.5





@dataclass
class ModelSpec:
    key: str
    display_name: str
    family: str
    image_size: Tuple[int, int]
    use_imagenet_norm: bool
    lr: float
    stage1_weight_path: str
    stage2_weight_path: str
    best_stage1_pipeline: List[str] = field(default_factory=list)
    best_stage2_pipeline: List[str] = field(default_factory=list)




MODEL_SPECS: Dict[str, ModelSpec] = {
    "densenet": ModelSpec(
        key="densenet",
        display_name="DenseNet121",
        family="CNN-based",
        image_size=(64, 64),
        use_imagenet_norm=False,
        lr=1e-4,
        stage1_weight_path=os.path.join(DENSENET_OUT_DIR, "stage1_densenet32_05.pt"),
        stage2_weight_path=os.path.join(DENSENET_OUT_DIR, "stage2_densenet32_05.pt"),
        # TODO：填入 DenseNet 單一模型 Greedy Search 得到的最佳 Stage1 增強組合
        best_stage1_pipeline=[],
        # DenseNet 單一模型 Greedy Search 選出的最佳 Stage2 增強組合
        best_stage2_pipeline=["noise"],
    ),
    "swin": ModelSpec(
        key="swin",
        display_name="Swin Transformer",
        family="Transformer-based",
        image_size=(224, 224),
        use_imagenet_norm=True,
        lr=5e-5,
        stage1_weight_path=os.path.join(SWIN_OUT_DIR, "stage1_imagenet_swin_tiny_src64_to224_lr5e5.pt"),
        stage2_weight_path=os.path.join(SWIN_OUT_DIR, "stage2_imagenet_swin_tiny_src64_to224_lr5e5.pt"),
        # TODO：填入 Swin 單一模型 Greedy Search 得到的最佳 Stage1 增強組合
        best_stage1_pipeline=[],
        # Swin 單一模型 Greedy Search 選出的最佳 Stage2 增強組合
        best_stage2_pipeline=["dilation", "sharpen"],
    ),
    "vgg11": ModelSpec(
        key="vgg11",
        display_name="Official CNN VGG11-BN-64",
        family="Official CNN-based VGG",
        image_size=CNN_IMAGE_SIZE,
        use_imagenet_norm=False,
        lr=CNN_LR,
        stage1_weight_path=os.path.join(CNN_OUT_DIR, "stage1_official_vgg11bn_64.pt"),
        stage2_weight_path=os.path.join(CNN_OUT_DIR, "stage2_official_vgg11bn_64.pt"),
        # TODO：填入 Official CNN VGG11-BN-64 單一模型 Greedy Search 得到的最佳 Stage1 增強組合
        best_stage1_pipeline=[],
        # Official CNN VGG11-BN-64 單一模型 Greedy Search 選出的最佳 Stage2 增強組合
        # 若你還沒有跑 CNN 的 greedy augmentation，可先維持 []。
        best_stage2_pipeline=['sharpen', 'noise'],
    ),
}


BACKBONES = ["vgg11", "densenet", "swin"]

print("\n[INFO] Model weight paths:")
for _k, _spec in MODEL_SPECS.items():
    print(f"  {_k:<8} Stage1: {_spec.stage1_weight_path} | exists={os.path.exists(_spec.stage1_weight_path)}")
    print(f"  {_k:<8} Stage2: {_spec.stage2_weight_path} | exists={os.path.exists(_spec.stage2_weight_path)}")

print("\n[INFO] Fixed augmentation pipelines:")
for _k, _spec in MODEL_SPECS.items():
    print(f"  {_k:<8} Stage1 pipeline: {_spec.best_stage1_pipeline}")
    print(f"  {_k:<8} Stage2 pipeline: {_spec.best_stage2_pipeline}")




# =========================================================
# 3. Utility：紀錄、存檔、評估圖表
# =========================================================
def log(msg: str):
    now = datetime.now().strftime("%H:%M:%S")
    print(f"[{now}] {msg}", flush=True)




def set_all_seeds(seed: int):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)




def make_unlocked_output_path(path: str) -> str:
    """Return a timestamped path when the original file is locked/opened by Excel."""
    folder = os.path.dirname(path)
    stem, ext = os.path.splitext(os.path.basename(path))
    ts = datetime.now().strftime("%Y%m%d_%H%M%S")
    return os.path.join(folder, f"{stem}_{ts}{ext}")


def round_float_columns(df: pd.DataFrame, ndigits: int = ROUND_OUTPUT_DECIMALS) -> pd.DataFrame:
    """將所有浮點數欄位四捨五入到指定小數位，主要用於 CSV 輸出。"""
    out = df.copy()
    float_cols = out.select_dtypes(include=["float", "float32", "float64"]).columns
    if len(float_cols) > 0:
        out[float_cols] = out[float_cols].round(ndigits)
    return out


def round_nested_floats(obj, ndigits: int = ROUND_OUTPUT_DECIMALS):
    """將 dict / list / tuple 中的 float 四捨五入，主要用於 JSON 輸出。"""
    if isinstance(obj, float):
        return round(obj, ndigits)
    if isinstance(obj, dict):
        return {k: round_nested_floats(v, ndigits) for k, v in obj.items()}
    if isinstance(obj, list):
        return [round_nested_floats(v, ndigits) for v in obj]
    if isinstance(obj, tuple):
        return tuple(round_nested_floats(v, ndigits) for v in obj)
    return obj


def save_csv(df: pd.DataFrame, path: str):
    """Save CSV safely.

    If Windows blocks writing because the CSV is opened by Excel or another program,
    the function saves to a timestamped filename instead of stopping the whole run.
    All floating-point columns are rounded to ROUND_OUTPUT_DECIMALS.
    """
    os.makedirs(os.path.dirname(path), exist_ok=True)
    df_to_save = round_float_columns(df, ROUND_OUTPUT_DECIMALS)
    try:
        df_to_save.to_csv(path, index=False, encoding="utf-8-sig")
        print(f"[INFO] Saved CSV: {path}")
    except PermissionError:
        alt_path = make_unlocked_output_path(path)
        df_to_save.to_csv(alt_path, index=False, encoding="utf-8-sig")
        print(f"[WARN] Permission denied when saving CSV: {path}")
        print("[WARN] The file is probably opened by Excel or another program.")
        print(f"[INFO] Saved CSV to alternative path instead: {alt_path}")


def save_json(data: dict, path: str):
    """Save JSON safely, with timestamp fallback if the target file is locked."""
    import json
    os.makedirs(os.path.dirname(path), exist_ok=True)
    try:
        with open(path, "w", encoding="utf-8") as f:
            json.dump(round_nested_floats(data, ROUND_OUTPUT_DECIMALS), f, indent=2, ensure_ascii=False)
        print(f"[INFO] Saved JSON: {path}")
    except PermissionError:
        alt_path = make_unlocked_output_path(path)
        with open(alt_path, "w", encoding="utf-8") as f:
            json.dump(round_nested_floats(data, ROUND_OUTPUT_DECIMALS), f, indent=2, ensure_ascii=False)
        print(f"[WARN] Permission denied when saving JSON: {path}")
        print("[WARN] The file is probably opened by another program.")
        print(f"[INFO] Saved JSON to alternative path instead: {alt_path}")




def plot_cm(cm, labels, title, normalize="true", show_counts=False, save_path=None):
    try:
        import matplotlib.pyplot as plt
    except Exception as e:
        print(f"[WARN] Skip confusion-matrix plot: {title}. Reason: {e}")
        return


    cm = np.asarray(cm, dtype=np.int64)
    labels = list(labels)


    if normalize == "true":
        row_sum = cm.sum(axis=1, keepdims=True)
        cm_show = np.divide(
            cm.astype(np.float64), row_sum,
            out=np.zeros_like(cm, dtype=np.float64),
            where=row_sum != 0,
        ) * 100.0
        colorbar_label = "Row percentage (%)"
        vmin, vmax = 0, 100
    else:
        cm_show = cm.astype(np.float64)
        colorbar_label = "Count"
        vmin, vmax = None, None


    fig_w = max(8, len(labels) * 0.9)
    fig_h = max(7, len(labels) * 0.75)
    plt.figure(figsize=(fig_w, fig_h))
    plt.imshow(cm_show, interpolation="nearest", cmap="Blues", vmin=vmin, vmax=vmax)
    plt.title(title)
    cbar = plt.colorbar()
    cbar.set_label(colorbar_label)
    plt.xticks(range(len(labels)), labels, rotation=45, ha="right")
    plt.yticks(range(len(labels)), labels)
    plt.xlabel("Predicted")
    plt.ylabel("True")


    for i in range(cm.shape[0]):
        row_total = cm[i].sum()
        for j in range(cm.shape[1]):
            count = int(cm[i, j])
            pct = (count / row_total * 100.0) if row_total > 0 else 0.0
            if normalize == "true":
                text = f"{count}\n{pct:.1f}%" if show_counts else f"{pct:.1f}%"
                color = "white" if cm_show[i, j] > 50 else "black"
            else:
                text = str(count)
                color = "white" if cm_show[i, j] > cm_show.max() * 0.5 else "black"
            plt.text(j, i, text, ha="center", va="center", color=color, fontsize=8)


    plt.tight_layout()
    if save_path is not None:
        os.makedirs(os.path.dirname(save_path), exist_ok=True)
        plt.savefig(save_path, dpi=200, bbox_inches="tight")
        print(f"[INFO] Saved confusion matrix: {save_path}")
    plt.show()




def evaluate_stage1_binary_report(true_folder, stage1_pred_name, title="Stage1 None vs Defect"):
    y_true_s1 = np.where(np.asarray(true_folder).astype(str) == NONE_CLASS_NAME, "None", "Defect")
    y_pred_s1 = np.asarray(stage1_pred_name).astype(str)
    labels_s1 = ["None", "Defect"]


    print("\n==============================")
    print(title)
    print("==============================")
    print(classification_report(
        y_true_s1, y_pred_s1,
        labels=labels_s1,
        target_names=labels_s1,
        digits=4,
        zero_division=0,
    ))
    cm = confusion_matrix(y_true_s1, y_pred_s1, labels=labels_s1)
    plot_cm(cm, labels_s1, title, normalize="true", show_counts=False)
    return cm




def evaluate_final_report(true_folder, final_pred, title="Final Two-stage Confusion Matrix"):
    final_labels = [NONE_CLASS_NAME] + DEFECT_CLASS_LIST
    print("\n==============================")
    print("Final Two-stage Classification Report")
    print("==============================")
    print(classification_report(
        true_folder, final_pred,
        labels=final_labels,
        target_names=final_labels,
        digits=4,
        zero_division=0,
    ))
    final_macro_f1 = f1_score(
        true_folder, final_pred,
        labels=final_labels,
        average="macro",
        zero_division=0,
    )
    print(f"Final macro-F1 = {final_macro_f1:.2f}")
    cm = confusion_matrix(true_folder, final_pred, labels=final_labels)
    plot_cm(cm, final_labels, title, normalize="true", show_counts=False)
    return final_macro_f1, cm




def stage1_false_negative_analysis(df_out, none_class_name="None"):
    print("\n==============================")
    print("Stage1 False-Negative Analysis (Defect -> None)")
    print("==============================")
    df_fn = df_out[
        (df_out["true_folder"].astype(str) != "") &
        (df_out["true_folder"].astype(str) != none_class_name) &
        (df_out["stage1_pred"].astype(str) == "None")
    ].copy()
    if len(df_fn) == 0:
        print("No defect samples were rejected as None by Stage1.")
        return
    print(f"Total Stage1 false negatives: {len(df_fn)}")
    print(df_fn["true_folder"].value_counts().sort_index())




# =========================================================
# 4. Data：讀取 samples、資料增強、Dataset、DataLoader
# =========================================================
def build_samples(root_dir: str, classes: List[str]):
    samples = []
    class_to_idx = {c: i for i, c in enumerate(classes)}
    for cls in classes:
        cls_dir = os.path.join(root_dir, cls)
        if not os.path.isdir(cls_dir):
            print(f"[WARN] Class folder not found: {cls_dir}")
            continue
        for fn in sorted(os.listdir(cls_dir)):
            fp = os.path.join(cls_dir, fn)
            if not os.path.isfile(fp):
                continue
            samples.append({
                "path": os.path.normpath(fp),
                "relpath": os.path.join(cls, fn),
                "class_name": cls,
                "class_idx": class_to_idx[cls],
            })
    return samples, class_to_idx




def ensure_nonempty_samples(samples, name):
    if len(samples) == 0:
        raise RuntimeError(f"{name} has no samples. Please check dataset path and class folders.")




# ---------- Wafer-specific augmentation ----------
def wafer_dilation(img):
    img_np = np.array(img)
    kernel = np.ones((2, 2), np.uint8)
    return Image.fromarray(cv2.dilate(img_np, kernel))




def wafer_erosion(img):
    img_np = np.array(img)
    kernel = np.ones((2, 2), np.uint8)
    return Image.fromarray(cv2.erode(img_np, kernel))




def wafer_sharpen(img):
    img_np = np.array(img)
    kernel = np.array([[0, -1, 0], [-1, 5, -1], [0, -1, 0]])
    return Image.fromarray(cv2.filter2D(img_np, -1, kernel))




def wafer_noise(img, rng):
    img_np = np.array(img)
    noise = rng.normal(5, 30, img_np.shape).astype(np.int16)
    img_np = np.clip(img_np.astype(np.int16) + noise, 0, 255).astype(np.uint8)
    return Image.fromarray(img_np)




def wafer_random_erasing(img, rng, erase_width_ratio=0.95, erase_height_ratio=0.05, safe_center_ratio=0.4):
    img_np = np.array(img)
    h, w, _ = img_np.shape
    erase_w = max(1, min(int(w * erase_width_ratio), w))
    erase_h = max(1, min(int(h * erase_height_ratio), h))
    center_x, center_y = w // 2, h // 2
    safe_radius_x = int(w * safe_center_ratio / 2)
    safe_radius_y = int(h * safe_center_ratio / 2)
    x0, y0 = 0, 0
    for _ in range(50):
        x0 = int(rng.integers(0, max(1, w - erase_w + 1)))
        y0 = int(rng.integers(0, max(1, h - erase_h + 1)))
        erase_x1 = x0 + erase_w
        erase_y1 = y0 + erase_h
        safe_x0 = center_x - safe_radius_x
        safe_x1 = center_x + safe_radius_x
        safe_y0 = center_y - safe_radius_y
        safe_y1 = center_y + safe_radius_y
        overlap_safe = not (
            erase_x1 <= safe_x0 or x0 >= safe_x1 or
            erase_y1 <= safe_y0 or y0 >= safe_y1
        )
        if not overlap_safe:
            break
    img_np[y0:y0 + erase_h, x0:x0 + erase_w, :] = 0
    return Image.fromarray(img_np)




def aug_apply_wafer(img_pil, op, rng):
    if op == "dilation":
        return wafer_dilation(img_pil)
    if op == "erosion":
        return wafer_erosion(img_pil)
    if op == "sharpen":
        return wafer_sharpen(img_pil)
    if op == "noise":
        return wafer_noise(img_pil, rng)
    if op == "erase":
        return wafer_random_erasing(img_pil, rng)
    return img_pil




def build_transform(spec: ModelSpec):
    items = [
        transforms.Resize(spec.image_size, interpolation=RESIZE_INTERPOLATION),
        transforms.ToTensor(),
    ]
    if spec.use_imagenet_norm:
        items.append(transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD))
    return transforms.Compose(items)




class WaferDataset(Dataset):
    def __init__(
        self,
        samples,
        transform,
        binary_none_name=None,
        defect_only=False,
        pipeline=None,
        apply_prob=0.7,
        seed=42,
    ):
        self.transform = transform
        self.binary_none_name = binary_none_name
        self.defect_only = defect_only
        self.pipeline = list(pipeline) if pipeline else []
        self.apply_prob = float(apply_prob)
        self.seed = int(seed)
        self.samples = [s for s in samples if (not defect_only or s["class_name"] != NONE_CLASS_NAME)]
        pipe_sig = "|".join(self.pipeline).encode("utf-8")
        self.pipe_hash = zlib.crc32(pipe_sig) & 0xFFFFFFFF
        self.defect_name_to_idx = {name: i for i, name in enumerate(DEFECT_CLASS_LIST)}


    def __len__(self):
        return len(self.samples)


    def _apply_pipeline(self, img, idx):
        if not self.pipeline:
            return img
        s = (self.seed * 1000003 + idx * 1009 + self.pipe_hash) & 0xFFFFFFFF
        rng = np.random.default_rng(s)
        out = img
        for op in self.pipeline:
            if rng.random() < self.apply_prob:
                out = aug_apply_wafer(out, op, rng)
        return out


    def __getitem__(self, idx):
        row = self.samples[idx]
        img = Image.open(row["path"]).convert("RGB")
        img = self._apply_pipeline(img, idx)
        x = self.transform(img)


        if self.binary_none_name is not None:
            y = 0 if row["class_name"] == self.binary_none_name else 1
        elif self.defect_only:
            y = self.defect_name_to_idx[row["class_name"]]
        else:
            y = row["class_idx"]


        meta = {
            "path": row["path"],
            "relpath": row["relpath"],
            "class_name": row["class_name"],
        }
        return x, y, meta




def build_loader(dataset, batch_size, shuffle=False, sampler=None):
    loader_kwargs = {}
    if NUM_WORKERS > 0:
        loader_kwargs["persistent_workers"] = True
        loader_kwargs["prefetch_factor"] = 2
    return DataLoader(
        dataset,
        batch_size=batch_size,
        shuffle=shuffle if sampler is None else False,
        sampler=sampler,
        num_workers=NUM_WORKERS,
        pin_memory=torch.cuda.is_available(),
        **loader_kwargs,
    )




# =========================================================
# 5. Models：Official CNN VGG11-BN / DenseNet / Swin
# =========================================================
class DenseNet64Classifier(nn.Module):
    def __init__(self, num_outputs, drop_rate=0.0):
        super().__init__()
        self.backbone = models.densenet121(weights=None, drop_rate=drop_rate)
        self.backbone.features.conv0 = nn.Conv2d(3, 64, kernel_size=3, stride=1, padding=1, bias=False)
        self.backbone.features.pool0 = nn.Identity()
        in_features = self.backbone.classifier.in_features
        self.backbone.classifier = nn.Linear(in_features, num_outputs)


    def forward(self, x):
        return self.backbone(x)




class SwinClassifierPT(nn.Module):
    def __init__(self, num_outputs, drop_rate=0.1):
        super().__init__()
        if timm is None:
            raise RuntimeError("Swin needs timm. Please run: pip install timm")
        self.backbone = timm.create_model(
            SWIN_MODEL_NAME,
            pretrained=USE_IMAGENET_PRETRAINED_SWIN,
            num_classes=num_outputs,
            drop_rate=drop_rate,
        )


    def forward(self, x):
        return self.backbone(x)




class OfficialCNNVGG11BN64Classifier(nn.Module):
    """torchvision 官方 CNN：VGG11-BN for 64x64 wafer map。

    使用 torchvision.models.vgg11_bn 作為 CNN backbone。
    VGG11-BN 是純 CNN 架構，主要由 convolution、batch normalization、ReLU
    與 max pooling 組成，不使用 ResNet residual connection，
    也不使用 Transformer attention 或 Mamba token mixing。

    本研究設定：
        - input: 64x64 RGB wafer map
        - pretrained: False，從零訓練
        - classifier: 改成較小的分類器，輸出 Stage1 或 Stage2 類別數

    注意：
        這個模型會讀取：
            stage1_official_vgg11bn_64.pt
            stage2_official_vgg11bn_64.pt
        不會讀取 Mamba 的權重，因為兩者架構不同，權重不能共用。
    """
    def __init__(self, num_outputs=1):
        super().__init__()

        if USE_IMAGENET_PRETRAINED_CNN:
            weights = models.VGG11_BN_Weights.DEFAULT
        else:
            weights = None

        self.backbone = models.vgg11_bn(weights=weights)

        # torchvision VGG 內建 AdaptiveAvgPool2d((7, 7))，
        # 因此 classifier 第一層輸入維度維持 512 * 7 * 7。
        self.backbone.classifier = nn.Sequential(
            nn.Linear(512 * 7 * 7, 512),
            nn.ReLU(inplace=True),
            nn.Dropout(CNN_DROPOUT),
            nn.Linear(512, 128),
            nn.ReLU(inplace=True),
            nn.Dropout(0.3),
            nn.Linear(128, num_outputs),
        )

    def forward(self, x):
        return self.backbone(x)



def build_model(backbone_key: str, num_outputs: int):
    if backbone_key == "vgg11":
        return OfficialCNNVGG11BN64Classifier(num_outputs=num_outputs).to(DEVICE)
    if backbone_key == "densenet":
        return DenseNet64Classifier(num_outputs=num_outputs, drop_rate=0.0).to(DEVICE)
    if backbone_key == "swin":
        return SwinClassifierPT(num_outputs=num_outputs, drop_rate=SWIN_DROP).to(DEVICE)
    raise ValueError(f"Unknown backbone_key: {backbone_key}")



def sanity_check_models():
    print("\n==============================")
    print("Sanity check: model input / output shape")
    print("==============================")
    for key in BACKBONES:
        spec = MODEL_SPECS[key]
        model = build_model(key, num_outputs=2)
        model.eval()
        dummy = torch.randn(2, 3, spec.image_size[0], spec.image_size[1], device=DEVICE)
        with torch.no_grad():
            out = model(dummy)
        print(f"{key:<8} input={tuple(dummy.shape)} output={tuple(out.shape)}")
        del model
        if torch.cuda.is_available():
            torch.cuda.empty_cache()




# =========================================================
# 6. Loss Weight：每個模型可用自己的設定
# =========================================================
def compute_stage1_pos_weight(samples, strength=1.0):
    n_none = sum(1 for s in samples if s["class_name"] == NONE_CLASS_NAME)
    n_def = sum(1 for s in samples if s["class_name"] != NONE_CLASS_NAME)
    base = n_none / max(n_def, 1)
    # DenseNet 使用 strength；其他模型 strength=1.0。
    pos_weight = base ** strength
    return torch.tensor([pos_weight], dtype=torch.float32).to(DEVICE)




def compute_stage2_class_weights(samples, strength=1.0):
    counts = np.zeros(len(DEFECT_CLASS_LIST), dtype=np.float64)
    defect_to_idx = {name: i for i, name in enumerate(DEFECT_CLASS_LIST)}
    for s in samples:
        if s["class_name"] != NONE_CLASS_NAME:
            counts[defect_to_idx[s["class_name"]]] += 1
    n_samples = counts.sum()
    k = len(DEFECT_CLASS_LIST)
    counts_safe = np.where(counts == 0, 1.0, counts)
    raw_w = n_samples / (k * counts_safe)
    weights = raw_w ** strength
    return torch.tensor(weights, dtype=torch.float32).to(DEVICE)




def get_loss_strength(backbone_key: str):
    if backbone_key == "densenet":
        return DENSENET_LOSS_WEIGHT_STRENGTH
    return 1.0




# =========================================================
# 7. Train / Eval：共用訓練與預測函式
# =========================================================
def safe_load_weights(model, weight_path: str) -> bool:
    if not os.path.exists(weight_path):
        return False
    try:
        try:
            state = torch.load(weight_path, map_location=DEVICE, weights_only=True)
        except TypeError:
            state = torch.load(weight_path, map_location=DEVICE)
        model.load_state_dict(state)
        log(f"Loaded weights: {weight_path}")
        return True
    except Exception as e:
        log(f"Failed to load weights: {weight_path}")
        log(f"Reason: {repr(e)}")
        return False




def save_weights(model, weight_path: str):
    os.makedirs(os.path.dirname(weight_path), exist_ok=True)
    torch.save(model.state_dict(), weight_path)
    log(f"Saved weights: {weight_path}")




def run_one_epoch(model, loader, criterion, optimizer=None, phase="train", epoch=None, epochs=None):
    is_train = optimizer is not None
    model.train(is_train)
    total_loss = 0.0
    total_correct = 0
    total_count = 0
    desc = f"{phase} epoch {epoch}/{epochs}" if epoch is not None else phase
    pbar = tqdm(loader, desc=desc, leave=False, mininterval=5)


    for xb, yb, _ in pbar:
        xb = xb.to(DEVICE, non_blocking=True)
        yb = yb.to(DEVICE, non_blocking=True)


        if is_train:
            optimizer.zero_grad(set_to_none=True)


        with torch.set_grad_enabled(is_train):
            logits = model(xb)
            if logits.ndim == 1 or logits.shape[-1] == 1:
                logits_flat = logits.view(-1)
                loss = criterion(logits_flat, yb.float())
                preds = (torch.sigmoid(logits_flat) >= 0.5).long()
            else:
                loss = criterion(logits, yb.long())
                preds = torch.argmax(logits, dim=1)


            if is_train:
                loss.backward()
                optimizer.step()


        total_loss += loss.item() * xb.size(0)
        total_correct += (preds == yb.long()).sum().item()
        total_count += xb.size(0)
        pbar.set_postfix({
            "loss": f"{total_loss / max(total_count, 1):.2f}",
            "acc": f"{total_correct / max(total_count, 1):.2f}",
        })


    return total_loss / max(total_count, 1), total_correct / max(total_count, 1)




def train_model_pt(model, train_loader, val_loader, criterion, lr, epochs, patience=4):
    optimizer = Adam(model.parameters(), lr=lr, weight_decay=WEIGHT_DECAY)
    scheduler = ReduceLROnPlateau(optimizer, mode="min", factor=0.5, patience=2, min_lr=1e-6)
    best_state = copy.deepcopy(model.state_dict())
    best_val_loss = np.inf
    wait = 0


    for ep in range(1, epochs + 1):
        tr_loss, tr_acc = run_one_epoch(model, train_loader, criterion, optimizer=optimizer, phase="train", epoch=ep, epochs=epochs)
        va_loss, va_acc = run_one_epoch(model, val_loader, criterion, optimizer=None, phase="val", epoch=ep, epochs=epochs)
        scheduler.step(va_loss)
        log(f"Epoch {ep:02d}/{epochs} | train_loss={tr_loss:.2f} train_acc={tr_acc:.2f} | val_loss={va_loss:.2f} val_acc={va_acc:.2f}")


        if va_loss < best_val_loss:
            best_val_loss = va_loss
            best_state = copy.deepcopy(model.state_dict())
            wait = 0
        else:
            wait += 1
        if wait >= patience:
            log("Early stopping triggered")
            break


    model.load_state_dict(best_state)
    return model




def build_stage1_loaders_for_spec(spec: ModelSpec, train_samples, val_samples, test_samples, train_pipeline=None):
    transform = build_transform(spec)
    train_ds = WaferDataset(train_samples, transform, binary_none_name=NONE_CLASS_NAME, pipeline=train_pipeline, apply_prob=APPLY_AUG_PROB, seed=SEED)
    val_ds = WaferDataset(val_samples, transform, binary_none_name=NONE_CLASS_NAME, pipeline=None, seed=SEED)
    test_ds = WaferDataset(test_samples, transform, binary_none_name=NONE_CLASS_NAME, pipeline=None, seed=SEED)
    return (
        build_loader(train_ds, BATCH_SIZE, shuffle=True),
        build_loader(val_ds, BATCH_SIZE, shuffle=False),
        build_loader(test_ds, BATCH_SIZE, shuffle=False),
    )




def build_stage2_loaders_for_spec(spec: ModelSpec, train_samples, val_samples, test_samples, train_pipeline=None):
    transform = build_transform(spec)
    train_ds = WaferDataset(train_samples, transform, defect_only=True, pipeline=train_pipeline, apply_prob=APPLY_AUG_PROB, seed=SEED)
    val_ds = WaferDataset(val_samples, transform, defect_only=True, pipeline=None, seed=SEED)
    # test_all 需要包含 None，因為 final two-stage 要依照原始 test 順序輸出。
    test_all_ds = WaferDataset(test_samples, transform, defect_only=False, pipeline=None, seed=SEED)
    test_def_ds = WaferDataset(test_samples, transform, defect_only=True, pipeline=None, seed=SEED)
    return (
        build_loader(train_ds, BATCH_SIZE, shuffle=True),
        build_loader(val_ds, BATCH_SIZE, shuffle=False),
        build_loader(test_all_ds, BATCH_SIZE, shuffle=False),
        build_loader(test_def_ds, BATCH_SIZE, shuffle=False),
    )




def ensure_stage1_model(backbone_key: str, train_samples, val_samples, test_samples):
    spec = MODEL_SPECS[backbone_key]
    log(f"Prepare Stage1 model: {spec.display_name}")
    model = build_model(backbone_key, num_outputs=1)


    loaded = False
    if not FORCE_RETRAIN_SINGLE_MODELS:
        loaded = safe_load_weights(model, spec.stage1_weight_path)


    if not loaded:
        if not TRAIN_SINGLE_MODEL_IF_MISSING:
            raise FileNotFoundError(f"Stage1 weight not found or incompatible: {spec.stage1_weight_path}")
        log(f"Train Stage1 with best pipeline: {spec.best_stage1_pipeline}")
        train_loader, val_loader, _ = build_stage1_loaders_for_spec(spec, train_samples, val_samples, test_samples, train_pipeline=spec.best_stage1_pipeline)
        strength = get_loss_strength(backbone_key)
        pos_weight = compute_stage1_pos_weight(train_samples, strength=strength)
        criterion = nn.BCEWithLogitsLoss(pos_weight=pos_weight)
        model = train_model_pt(model, train_loader, val_loader, criterion, lr=spec.lr, epochs=EPOCHS_STAGE1, patience=4)
        save_weights(model, spec.stage1_weight_path)


    model.eval()
    return model




def ensure_stage2_model(backbone_key: str, train_samples, val_samples, test_samples):
    spec = MODEL_SPECS[backbone_key]
    log(f"Prepare Stage2 model: {spec.display_name}")
    model = build_model(backbone_key, num_outputs=len(DEFECT_CLASS_LIST))


    loaded = False
    if not FORCE_RETRAIN_SINGLE_MODELS:
        loaded = safe_load_weights(model, spec.stage2_weight_path)


    if not loaded:
        if not TRAIN_SINGLE_MODEL_IF_MISSING:
            raise FileNotFoundError(f"Stage2 weight not found or incompatible: {spec.stage2_weight_path}")
        log(f"Train Stage2 with best pipeline: {spec.best_stage2_pipeline}")
        train_loader, val_loader, _, _ = build_stage2_loaders_for_spec(spec, train_samples, val_samples, test_samples, train_pipeline=spec.best_stage2_pipeline)
        strength = get_loss_strength(backbone_key)
        class_weights = compute_stage2_class_weights(train_samples, strength=strength)
        criterion = nn.CrossEntropyLoss(weight=class_weights)
        model = train_model_pt(model, train_loader, val_loader, criterion, lr=spec.lr, epochs=EPOCHS_STAGE2, patience=5)
        save_weights(model, spec.stage2_weight_path)


    model.eval()
    return model




@torch.no_grad()
def predict_stage1_prob(model, loader):
    model.eval()
    probs = []
    metas = []
    y_true = []
    for xb, yb, mb in tqdm(loader, desc="predict Stage1", leave=False, mininterval=5):
        xb = xb.to(DEVICE, non_blocking=True)
        logits = model(xb).view(-1)
        p = torch.sigmoid(logits)
        probs.append(p.detach().cpu().numpy())
        y_true.extend(yb.numpy().tolist())
        metas.extend([{k: mb[k][i] for k in mb} for i in range(len(mb["path"]))])
    return np.concatenate(probs, axis=0), np.asarray(y_true, dtype=np.int64), metas




@torch.no_grad()
def predict_stage2_prob(model, loader):
    model.eval()
    probs = []
    metas = []
    y_true = []
    for xb, yb, mb in tqdm(loader, desc="predict Stage2", leave=False, mininterval=5):
        xb = xb.to(DEVICE, non_blocking=True)
        logits = model(xb)
        p = torch.softmax(logits, dim=1)
        probs.append(p.detach().cpu().numpy())
        y_true.extend(yb.numpy().tolist())
        metas.extend([{k: mb[k][i] for k in mb} for i in range(len(mb["path"]))])
    return np.concatenate(probs, axis=0), np.asarray(y_true, dtype=np.int64), metas




# =========================================================
# 8. Greedy-Based Selection Algorithm：使用已訓練 .pt 權重做模型集合選取
# =========================================================
# 文字敘述對應：
#
# Input:
#   多種已訓練完成的深度學習模型 M1, M2, ..., Mn。
#   本研究中為 Official CNN VGG11-BN-64、DenseNet、Swin Transformer。
#
# Output:
#   被選取的模型集合 S。
#
# Step 1:
#   根據各模型在 validation set 上的個別 F1 ratio 由大到小排序，
#   並初始化選取模型集合 S = 空集合。
#
# Step 2:
#   依排序選擇下一個 F1 ratio 最高且尚未選擇的模型 Mi，
#   暫時加入模型集合 S。
#
# Step 3:
#   使用目前模型集合 S 進行 soft voting 整合測試：
#       Stage1：平均各模型輸出的 Defect probability。
#       Stage2：平均各模型輸出的 8 類 softmax probability。
#
#   若新模型集合的 F1 ratio >= 上一次模型集合的 F1 ratio：
#       保留模型 Mi，若仍有未選模型則回到 Step 2。
#       本研究比較時只看到小數點第二位；若兩者小數點第二位相同，
#       視為新模型集合沒有下降，因此也允許加入。
#   否則：
#       將模型 Mi 從 S 中移除，並停止搜尋。
#
# Step 4:
#   輸出目前模型集合 S，作為該階段的最終模型集合。
#
# 注意：
#   這裡的 voting 採用 soft voting，也就是平均模型輸出的 probability。
#   原因是 Stage1 需要 threshold 搜尋，Stage2 需要類別機率整合；
#   使用 soft voting 比 hard voting 能保留模型信心分數。
# =========================================================


def normalize_weights(keys: List[str], raw_weights: Optional[Dict[str, float]] = None):
    """將模型權重正規化。預設每個模型等權重。"""
    if raw_weights is None:
        raw_weights = {k: 1.0 for k in keys}
    s = sum(float(raw_weights[k]) for k in keys)
    return {k: float(raw_weights[k]) / max(s, 1e-8) for k in keys}


def f1_ratio_for_comparison(value: float, ndigits: int = 2) -> float:
    """
    Greedy 模型加入條件使用的 F1 ratio 比較值。

    本研究在 Greedy-Based Selection 中只比較到小數點第二位：
        例如 0.844 與 0.842 皆視為 0.84。

    因此，若新模型集合與上一個模型集合在小數點第二位相同，
    就視為沒有下降，允許新模型加入。

    注意：這裡使用格式化到小數點第二位後再轉回 float，
    與表格或報告中顯示的 F1 ratio 一致。
    """
    return float(f"{float(value):.{ndigits}f}")


def is_f1_not_worse_by_2dec(new_f1: float, previous_f1: Optional[float]) -> bool:
    """
    判斷新模型集合是否可以保留。

    條件：
    1. 若尚未有 previous_f1，第一個模型一定保留。
    2. 若 new_f1 與 previous_f1 四捨五入到小數點第二位後相同，視為 new_f1 >= previous_f1。
    3. 若 new_f1 到小數點第二位後大於 previous_f1，也保留。
    """
    if previous_f1 is None:
        return True
    return f1_ratio_for_comparison(new_f1) >= f1_ratio_for_comparison(previous_f1)


def ensemble_stage1_prob(prob_map: Dict[str, np.ndarray], selected: List[str], weights: Dict[str, float]):
    """Stage1 soft voting：平均選取模型的 Defect probability。"""
    out = np.zeros_like(prob_map[selected[0]], dtype=np.float64)
    for k in selected:
        out += weights[k] * prob_map[k]
    return out


def ensemble_stage2_prob(prob_map: Dict[str, np.ndarray], selected: List[str], weights: Dict[str, float]):
    """Stage2 soft voting：平均選取模型的 8 類 softmax probability。"""
    out = np.zeros_like(prob_map[selected[0]], dtype=np.float64)
    for k in selected:
        out += weights[k] * prob_map[k]
    return out


def tune_binary_threshold_from_probs(y_true, y_prob, threshold_grid=None):
    """
    Stage1 threshold search。

    對同一組模型集合產生的 ensemble probability，
    嘗試多個 threshold，選出 F1 ratio 最高的 threshold。
    若 F1 相同，則選 false negative 較少者；再相同則選 recall 較高者。
    """
    if threshold_grid is None:
        threshold_grid = THRESH_GRID

    best = None
    rows = []

    for th in threshold_grid:
        y_pred = (y_prob >= th).astype(int)

        tp = np.sum((y_true == 1) & (y_pred == 1))
        fn = np.sum((y_true == 1) & (y_pred == 0))
        fp = np.sum((y_true == 0) & (y_pred == 1))

        recall = tp / (tp + fn + 1e-8)
        precision = tp / (tp + fp + 1e-8)
        f1 = 2 * precision * recall / (precision + recall + 1e-8)

        row = {
            "threshold": round(float(th), ROUND_OUTPUT_DECIMALS),
            "fn": int(fn),
            "recall": round(float(recall), ROUND_OUTPUT_DECIMALS),
            "precision": round(float(precision), ROUND_OUTPUT_DECIMALS),
            "f1_ratio": round(float(f1), ROUND_OUTPUT_DECIMALS),
            "f1": round(float(f1), ROUND_OUTPUT_DECIMALS),  # 保留 f1 欄位，方便與原本輸出相容
        }
        rows.append(row)

        if best is None:
            best = row
        else:
            cur_key = (-row["f1_ratio"], row["fn"], -row["recall"])
            best_key = (-best["f1_ratio"], best["fn"], -best["recall"])
            if cur_key < best_key:
                best = row

    return best, pd.DataFrame(rows)


def evaluate_stage1_selected_models(prob_map: Dict[str, np.ndarray], y_true: np.ndarray, selected: List[str]):
    """評估 Stage1 某個模型集合，回傳最佳 threshold 與 F1 ratio。"""
    weights = normalize_weights(selected)
    ens_prob = ensemble_stage1_prob(prob_map, selected, weights)
    th_info, _ = tune_binary_threshold_from_probs(y_true.astype(int), ens_prob, THRESH_GRID)
    return {
        "selected_models": "+".join(selected),
        "n_models": len(selected),
        "threshold": th_info["threshold"],
        "fn": th_info["fn"],
        "recall": th_info["recall"],
        "precision": th_info["precision"],
        "f1_ratio": th_info["f1_ratio"],
        "f1": th_info["f1_ratio"],
        **{k: 1 if k in selected else 0 for k in BACKBONES},
    }


def evaluate_stage2_selected_models(prob_map: Dict[str, np.ndarray], y_true_idx: np.ndarray, selected: List[str]):
    """評估 Stage2 某個模型集合，回傳 macro-F1 ratio 與 accuracy。"""
    weights = normalize_weights(selected)
    ens_prob = ensemble_stage2_prob(prob_map, selected, weights)
    pred_idx = np.argmax(ens_prob, axis=1)
    macro_f1 = f1_score(
        y_true_idx,
        pred_idx,
        labels=list(range(len(DEFECT_CLASS_LIST))),
        average="macro",
        zero_division=0,
    )
    acc = accuracy_score(y_true_idx, pred_idx)
    return {
        "selected_models": "+".join(selected),
        "n_models": len(selected),
        "macro_f1": round(float(macro_f1), ROUND_OUTPUT_DECIMALS),
        "f1_ratio": round(float(macro_f1), ROUND_OUTPUT_DECIMALS),
        "acc": round(float(acc), ROUND_OUTPUT_DECIMALS),
        **{k: 1 if k in selected else 0 for k in BACKBONES},
    }


def greedy_based_selection_stage1(prob_map: Dict[str, np.ndarray], y_true: np.ndarray):
    """
    Stage1 Greedy-Based Selection Algorithm。

    Step 1：
        先計算每個單一模型的 Stage1 F1 ratio，並由大到小排序。
    Step 2：
        依序將下一個模型加入目前集合。
    Step 3：
        使用 soft voting 評估新集合，若 F1 ratio 沒有下降則保留，
        否則移除該模型並停止。
    Step 4：
        輸出最後保留的模型集合。
    """
    print("\n==============================")
    print("Greedy-Based Selection: Stage1 Model Set")
    print("==============================")

    # Step 1. 個別模型排序
    single_rows = []
    for key in BACKBONES:
        row = evaluate_stage1_selected_models(prob_map, y_true, [key])
        row["step"] = "single_model_ranking"
        row["candidate_model"] = key
        row["decision"] = "ranking_only"
        single_rows.append(row)

    df_single = pd.DataFrame(single_rows).sort_values(
        ["f1_ratio", "recall", "fn"],
        ascending=[False, False, True],
    ).reset_index(drop=True)

    sorted_models = df_single["selected_models"].tolist()

    print("[Stage1] Individual model ranking by F1 ratio:")
    for _, r in df_single.iterrows():
        print(
            f"  {r['selected_models']:<10} "
            f"F1={r['f1_ratio']:.2f} recall={r['recall']:.2f} "
            f"fn={int(r['fn'])} th={r['threshold']:.2f}"
        )

    # Step 2-4. 逐步加入模型
    selected = []
    best_row = None
    rows = single_rows[:]
    previous_f1 = None

    for step_idx, model_key in enumerate(sorted_models, start=1):
        trial_selected = selected + [model_key]
        trial_row = evaluate_stage1_selected_models(prob_map, y_true, trial_selected)
        trial_row["step"] = f"greedy_step_{step_idx}"
        trial_row["candidate_model"] = model_key

        trial_row["f1_ratio_cmp2"] = f1_ratio_for_comparison(trial_row["f1_ratio"])
        trial_row["previous_f1_ratio"] = previous_f1
        trial_row["previous_f1_ratio_cmp2"] = (
            np.nan if previous_f1 is None else f1_ratio_for_comparison(previous_f1)
        )

        if is_f1_not_worse_by_2dec(trial_row["f1_ratio"], previous_f1):
            selected = trial_selected
            previous_f1 = trial_row["f1_ratio"]
            best_row = trial_row.copy()
            trial_row["decision"] = "keep"
            trial_row["decision_rule"] = "keep_if_f1_rounded_2dec_not_worse"
            print(
                f"[Stage1 Step {step_idx}] add={model_key:<8} KEEP | "
                f"models={trial_row['selected_models']:<24} "
                f"F1={trial_row['f1_ratio']:.2f} "
                f"F1_cmp2={trial_row['f1_ratio_cmp2']:.2f} "
                f"prev_cmp2={trial_row['previous_f1_ratio_cmp2'] if previous_f1 is not None else 'None'} "
                f"fn={trial_row['fn']} recall={trial_row['recall']:.2f}"
            )
        else:
            trial_row["decision"] = "remove_and_stop"
            trial_row["decision_rule"] = "remove_if_f1_rounded_2dec_lower"
            print(
                f"[Stage1 Step {step_idx}] add={model_key:<8} REMOVE & STOP | "
                f"models={trial_row['selected_models']:<24} "
                f"F1={trial_row['f1_ratio']:.2f} "
                f"F1_cmp2={trial_row['f1_ratio_cmp2']:.2f} "
                f"< previous_cmp2={f1_ratio_for_comparison(previous_f1):.2f}"
            )
            rows.append(trial_row)
            break

        rows.append(trial_row)

    if best_row is None:
        # 理論上不會發生，保險處理：取單一模型第一名
        best_row = df_single.iloc[0].to_dict()
        selected = [best_row["selected_models"]]

    best_row["selected_models"] = "+".join(selected)
    best_row["n_models"] = len(selected)
    for k in BACKBONES:
        best_row[k] = 1 if k in selected else 0

    df = pd.DataFrame(rows)
    sort_cols = ["step", "selected_models"]
    existing_sort_cols = [c for c in sort_cols if c in df.columns]
    if existing_sort_cols:
        df = df.reset_index(drop=True)

    return best_row, df


def greedy_based_selection_stage2(prob_map: Dict[str, np.ndarray], y_true_idx: np.ndarray):
    """
    Stage2 Greedy-Based Selection Algorithm。

    Stage2 的 F1 ratio 使用 macro-F1。
    因為 Stage2 是 8 類缺陷分類，macro-F1 能降低類別不平衡對評估的影響。
    """
    print("\n==============================")
    print("Greedy-Based Selection: Stage2 Model Set")
    print("==============================")

    # Step 1. 個別模型排序
    single_rows = []
    for key in BACKBONES:
        row = evaluate_stage2_selected_models(prob_map, y_true_idx, [key])
        row["step"] = "single_model_ranking"
        row["candidate_model"] = key
        row["decision"] = "ranking_only"
        single_rows.append(row)

    df_single = pd.DataFrame(single_rows).sort_values(
        ["f1_ratio", "acc"],
        ascending=[False, False],
    ).reset_index(drop=True)

    sorted_models = df_single["selected_models"].tolist()

    print("[Stage2] Individual model ranking by macro-F1 ratio:")
    for _, r in df_single.iterrows():
        print(
            f"  {r['selected_models']:<10} "
            f"macro-F1={r['f1_ratio']:.2f} acc={r['acc']:.2f}"
        )

    # Step 2-4. 逐步加入模型
    selected = []
    best_row = None
    rows = single_rows[:]
    previous_f1 = None

    for step_idx, model_key in enumerate(sorted_models, start=1):
        trial_selected = selected + [model_key]
        trial_row = evaluate_stage2_selected_models(prob_map, y_true_idx, trial_selected)
        trial_row["step"] = f"greedy_step_{step_idx}"
        trial_row["candidate_model"] = model_key

        trial_row["f1_ratio_cmp2"] = f1_ratio_for_comparison(trial_row["f1_ratio"])
        trial_row["previous_f1_ratio"] = previous_f1
        trial_row["previous_f1_ratio_cmp2"] = (
            np.nan if previous_f1 is None else f1_ratio_for_comparison(previous_f1)
        )

        if is_f1_not_worse_by_2dec(trial_row["f1_ratio"], previous_f1):
            selected = trial_selected
            previous_f1 = trial_row["f1_ratio"]
            best_row = trial_row.copy()
            trial_row["decision"] = "keep"
            trial_row["decision_rule"] = "keep_if_f1_rounded_2dec_not_worse"
            print(
                f"[Stage2 Step {step_idx}] add={model_key:<8} KEEP | "
                f"models={trial_row['selected_models']:<24} "
                f"macro-F1={trial_row['f1_ratio']:.2f} "
                f"F1_cmp2={trial_row['f1_ratio_cmp2']:.2f} "
                f"prev_cmp2={trial_row['previous_f1_ratio_cmp2'] if previous_f1 is not None else 'None'} "
                f"acc={trial_row['acc']:.2f}"
            )
        else:
            trial_row["decision"] = "remove_and_stop"
            trial_row["decision_rule"] = "remove_if_f1_rounded_2dec_lower"
            print(
                f"[Stage2 Step {step_idx}] add={model_key:<8} REMOVE & STOP | "
                f"models={trial_row['selected_models']:<24} "
                f"macro-F1={trial_row['f1_ratio']:.2f} "
                f"F1_cmp2={trial_row['f1_ratio_cmp2']:.2f} "
                f"< previous_cmp2={f1_ratio_for_comparison(previous_f1):.2f}"
            )
            rows.append(trial_row)
            break

        rows.append(trial_row)

    if best_row is None:
        # 理論上不會發生，保險處理：取單一模型第一名
        best_row = df_single.iloc[0].to_dict()
        selected = [best_row["selected_models"]]

    best_row["selected_models"] = "+".join(selected)
    best_row["n_models"] = len(selected)
    for k in BACKBONES:
        best_row[k] = 1 if k in selected else 0

    df = pd.DataFrame(rows).reset_index(drop=True)
    return best_row, df


def analyze_weak_classes_against_best_single(
    df_out: pd.DataFrame,
    best_single_model: str,
    selected_stage2: List[str],
    top_k: int = 3,
):
    """
    挑選目前最佳單一模型表現較差的幾個 defect 類別，
    比較 Greedy-Based Selection 選出的模型集合是否優於單一模型。

    分析邏輯：
    1. 以 Stage2 validation 表現最佳的單一模型作為 baseline。
    2. 在 test set 中計算該單一模型於各 defect 類別的 F1。
    3. 挑選 F1 最低的 top_k 個類別作為弱勢類別。
    4. 比較：
       - best single model 的 stage2 prediction
       - greedy-selected model set 的 stage2 prediction
       - final two-stage prediction
    """
    if best_single_model not in BACKBONES:
        raise ValueError(f"Unknown best_single_model: {best_single_model}")

    single_pred_col = f"stage2_pred_{best_single_model}"
    if single_pred_col not in df_out.columns:
        raise ValueError(f"df_out missing column: {single_pred_col}")

    df_def = df_out[df_out["true_folder"].astype(str).isin(DEFECT_CLASS_LIST)].copy()
    if len(df_def) == 0:
        raise ValueError("No defect samples found for weak-class analysis.")

    y_true = df_def["true_folder"].astype(str).to_numpy()
    y_single = df_def[single_pred_col].astype(str).to_numpy()
    y_greedy_stage2 = df_def["stage2_pred"].astype(str).to_numpy()
    y_final = df_def["final_pred"].astype(str).to_numpy()

    # 單一模型各類別 F1
    p_single, r_single, f_single, support = precision_recall_fscore_support(
        y_true,
        y_single,
        labels=DEFECT_CLASS_LIST,
        zero_division=0,
    )
    p_greedy, r_greedy, f_greedy, _ = precision_recall_fscore_support(
        y_true,
        y_greedy_stage2,
        labels=DEFECT_CLASS_LIST,
        zero_division=0,
    )
    p_final, r_final, f_final, _ = precision_recall_fscore_support(
        y_true,
        y_final,
        labels=DEFECT_CLASS_LIST,
        zero_division=0,
    )

    per_class = pd.DataFrame({
        "class_name": DEFECT_CLASS_LIST,
        "support": support,
        "best_single_model": best_single_model,
        "best_single_precision": p_single,
        "best_single_recall": r_single,
        "best_single_f1": f_single,
        "greedy_selected_models": "+".join(selected_stage2),
        "greedy_stage2_precision": p_greedy,
        "greedy_stage2_recall": r_greedy,
        "greedy_stage2_f1": f_greedy,
        "final_two_stage_precision": p_final,
        "final_two_stage_recall": r_final,
        "final_two_stage_f1": f_final,
    })

    weak_classes = (
        per_class[per_class["support"] > 0]
        .sort_values(["best_single_f1", "support"], ascending=[True, False])
        .head(top_k)["class_name"]
        .tolist()
    )

    df_weak = df_def[df_def["true_folder"].astype(str).isin(weak_classes)].copy()

    summary_rows = []
    for name, pred_col in [
        (f"best_single_{best_single_model}", single_pred_col),
        ("greedy_selected_stage2", "stage2_pred"),
        ("final_two_stage", "final_pred"),
    ]:
        y_w_true = df_weak["true_folder"].astype(str).to_numpy()
        y_w_pred = df_weak[pred_col].astype(str).to_numpy()
        summary_rows.append({
            "analysis_type": "weak_classes_summary",
            "model_or_method": name,
            "weak_classes": "+".join(weak_classes),
            "support": int(len(df_weak)),
            "accuracy": round(float(accuracy_score(y_w_true, y_w_pred)), ROUND_OUTPUT_DECIMALS),
            "macro_f1": round(float(f1_score(
                y_w_true,
                y_w_pred,
                labels=weak_classes,
                average="macro",
                zero_division=0,
            )), ROUND_OUTPUT_DECIMALS),
        })

    df_summary = pd.DataFrame(summary_rows)

    per_class["analysis_type"] = "per_defect_class"
    per_class["is_weak_class"] = per_class["class_name"].isin(weak_classes).astype(int)

    # 合併輸出：前半是每類別，後半是弱勢類別 summary。
    df_export = pd.concat(
        [
            per_class,
            df_summary,
        ],
        ignore_index=True,
        sort=False,
    )

    print("\n==============================")
    print("Weak-class analysis: Greedy selection vs best single model")
    print("==============================")
    print(f"Best single Stage2 model = {best_single_model}")
    print(f"Greedy-selected Stage2 models = {selected_stage2}")
    print(f"Weak classes selected = {weak_classes}")
    print(df_summary)

    save_csv(df_export, OUT_WEAK_CLASS_CSV)
    return df_export



# =========================================================
# 9. Sample60：固定樣本預測差異分析
# =========================================================
def build_or_load_sample60(df_out, force_rebuild=False):
    if (not force_rebuild) and os.path.exists(OUT_SAMPLE60_CSV):
        df_pick = pd.read_csv(OUT_SAMPLE60_CSV, encoding="utf-8-sig")
        print(f"[INFO] Loaded cached sample60: {OUT_SAMPLE60_CSV}")
        return df_pick


    os.makedirs(SAMPLE60_DIR, exist_ok=True)
    df_eval = df_out[df_out["true_folder"].astype(str) != ""].copy()
    if len(df_eval) == 0:
        raise ValueError("No labeled samples in TEST.")


    rng = np.random.default_rng(SEED)
    n_total = 60
    picked_rows = []


    # 先每類至少抽一張，再補滿 60 張。
    for cls in sorted(df_eval["true_folder"].astype(str).unique().tolist()):
        df_cls = df_eval[df_eval["true_folder"].astype(str) == cls]
        if len(df_cls) == 0:
            continue
        idx = rng.integers(0, len(df_cls))
        picked_rows.append(df_cls.iloc[int(idx)])


    picked_paths = set([str(r["path"]) for r in picked_rows])
    remaining_need = n_total - len(picked_rows)
    if remaining_need > 0:
        df_remain = df_eval[~df_eval["path"].astype(str).isin(picked_paths)].copy()
        extra_idx = rng.choice(len(df_remain), size=min(remaining_need, len(df_remain)), replace=False)
        for _, r in df_remain.iloc[extra_idx].iterrows():
            picked_rows.append(r)


    df_pick = pd.DataFrame(picked_rows).reset_index(drop=True)
    df_pick["correct_1or0"] = (df_pick["final_pred"].astype(str) == df_pick["true_folder"].astype(str)).astype(int)


    copied_paths = []
    for i, row in df_pick.iterrows():
        src = str(row["path"])
        cls = str(row["true_folder"])
        base = os.path.basename(src)
        dst_dir = os.path.join(SAMPLE60_DIR, cls)
        os.makedirs(dst_dir, exist_ok=True)
        dst = os.path.join(dst_dir, base)
        if os.path.exists(dst):
            name, ext = os.path.splitext(base)
            dst = os.path.join(dst_dir, f"{name}__idx{i:03d}{ext}")
        try:
            import shutil
            shutil.copy2(src, dst)
        except Exception as e:
            print(f"[WARN] Copy failed: {src} -> {dst}. Reason: {e}")
        copied_paths.append(dst)


    df_pick["copied_path"] = copied_paths
    save_csv(df_pick, OUT_SAMPLE60_CSV)


    print("\n==============================")
    print("Sample60 Summary")
    print("==============================")
    print(df_pick["correct_1or0"].value_counts(dropna=False))
    print(f"Saved sample60 images to: {SAMPLE60_DIR}")
    return df_pick




# =========================================================
# 10. Main：完整執行流程
# =========================================================
def main():
    set_all_seeds(SEED)


    # ---------- 10.1 讀取 train / val / test samples ----------
    log("Build sample lists.")
    train_samples, _ = build_samples(TRAIN_DIR, CLASS_LIST)
    val_samples, _ = build_samples(VAL_DIR, CLASS_LIST)
    test_samples, _ = build_samples(TEST_DIR, CLASS_LIST)
    ensure_nonempty_samples(train_samples, "train_samples")
    ensure_nonempty_samples(val_samples, "val_samples")
    ensure_nonempty_samples(test_samples, "test_samples")
    print(f"[INFO] train samples = {len(train_samples)}")
    print(f"[INFO] val samples   = {len(val_samples)}")
    print(f"[INFO] test samples  = {len(test_samples)}")


    # ---------- 10.2 檢查模型輸入 / 輸出 ----------
    sanity_check_models()


    # ---------- 10.3 載入或訓練三個單一模型的 Stage1 / Stage2 權重 ----------
    stage1_models = {}
    stage2_models = {}
    for key in BACKBONES:
        stage1_models[key] = ensure_stage1_model(key, train_samples, val_samples, test_samples)
        stage2_models[key] = ensure_stage2_model(key, train_samples, val_samples, test_samples)


    # ---------- 10.4 取得 validation predictions，用來做 Greedy model combination ----------
    log("Collect validation predictions for greedy model search.")
    val_stage1_prob_map = {}
    val_stage1_true = None
    for key in BACKBONES:
        spec = MODEL_SPECS[key]
        _, val_bin_loader, _ = build_stage1_loaders_for_spec(spec, train_samples, val_samples, test_samples, train_pipeline=None)
        prob, y_true, _ = predict_stage1_prob(stage1_models[key], val_bin_loader)
        val_stage1_prob_map[key] = prob
        if val_stage1_true is None:
            val_stage1_true = y_true


    val_stage2_prob_map = {}
    val_stage2_true = None
    for key in BACKBONES:
        spec = MODEL_SPECS[key]
        _, val_def_loader, _, _ = build_stage2_loaders_for_spec(spec, train_samples, val_samples, test_samples, train_pipeline=None)
        prob, y_true, _ = predict_stage2_prob(stage2_models[key], val_def_loader)
        val_stage2_prob_map[key] = prob
        if val_stage2_true is None:
            val_stage2_true = y_true


    # ---------- 10.5 Stage1 Greedy-Based Selection：選模型集合與 threshold ----------
    best_s1, df_s1 = greedy_based_selection_stage1(val_stage1_prob_map, val_stage1_true)
    save_csv(df_s1, GREEDY_STAGE1_MODELS_CSV)
    selected_stage1 = [k for k in BACKBONES if best_s1[k] == 1]
    stage1_weights = normalize_weights(selected_stage1)
    thresh_stage1 = float(best_s1["threshold"])
    print("\n[Selected Stage1 models]", selected_stage1)
    print("[Selected Stage1 weights]", stage1_weights)
    print("[Selected Stage1 threshold]", thresh_stage1)


    # ---------- 10.6 Stage2 Greedy-Based Selection：選模型集合 ----------
    best_s2, df_s2 = greedy_based_selection_stage2(val_stage2_prob_map, val_stage2_true)
    save_csv(df_s2, GREEDY_STAGE2_MODELS_CSV)
    selected_stage2 = [k for k in BACKBONES if best_s2[k] == 1]
    stage2_weights = normalize_weights(selected_stage2)

    # 找出 Stage2 validation 表現第一名的單一模型，供後續弱勢類別比較使用。
    df_s2_single = df_s2[df_s2["n_models"] == 1].copy()
    df_s2_single = df_s2_single.sort_values(["f1_ratio", "acc"], ascending=[False, False])
    best_single_stage2_model = str(df_s2_single.iloc[0]["selected_models"])
    print("\n[Selected Stage2 models]", selected_stage2)
    print("[Selected Stage2 weights]", stage2_weights)


    # ---------- 10.7 TEST：收集 Stage1 prediction ----------
    log("Collect TEST predictions.")
    test_stage1_prob_map = {}
    test_stage1_metas = None
    for key in BACKBONES:
        spec = MODEL_SPECS[key]
        _, _, test_bin_loader = build_stage1_loaders_for_spec(spec, train_samples, val_samples, test_samples, train_pipeline=None)
        prob, _, metas = predict_stage1_prob(stage1_models[key], test_bin_loader)
        test_stage1_prob_map[key] = prob
        if test_stage1_metas is None:
            test_stage1_metas = metas


    stage1_test_ens_prob = ensemble_stage1_prob(test_stage1_prob_map, selected_stage1, stage1_weights)
    stage1_pred_bin = (stage1_test_ens_prob >= thresh_stage1).astype(int)
    stage1_pred_name = np.where(stage1_pred_bin == 1, "Defect", "None")


    # ---------- 10.8 TEST：收集 Stage2 prediction ----------
    test_stage2_prob_map = {}
    test_stage2_metas = None
    for key in BACKBONES:
        spec = MODEL_SPECS[key]
        _, _, test_all_loader, _ = build_stage2_loaders_for_spec(spec, train_samples, val_samples, test_samples, train_pipeline=None)
        prob, _, metas = predict_stage2_prob(stage2_models[key], test_all_loader)
        test_stage2_prob_map[key] = prob
        if test_stage2_metas is None:
            test_stage2_metas = metas


    stage2_test_ens_prob = ensemble_stage2_prob(test_stage2_prob_map, selected_stage2, stage2_weights)
    stage2_pred_idx = np.argmax(stage2_test_ens_prob, axis=1)
    stage2_pred_name = np.array([DEFECT_CLASS_LIST[i] for i in stage2_pred_idx])
    stage2_conf = np.max(stage2_test_ens_prob, axis=1)


    # ---------- 10.9 Two-stage final decision ----------
    true_folder = np.array([m["class_name"] for m in test_stage2_metas])
    final_pred = np.where(stage1_pred_bin == 0, NONE_CLASS_NAME, stage2_pred_name)


    df_out = pd.DataFrame({
        "path": [m["path"] for m in test_stage2_metas],
        "relpath": [m["relpath"] for m in test_stage2_metas],
        "file": [os.path.basename(m["path"]) for m in test_stage2_metas],
        "true_folder": true_folder,
        "stage1_prob_ensemble": stage1_test_ens_prob,
        "stage1_pred": stage1_pred_name,
        "stage2_pred": stage2_pred_name,
        "stage2_conf": stage2_conf,
        "final_pred": final_pred,
    })
    for key in BACKBONES:
        df_out[f"stage1_prob_{key}"] = test_stage1_prob_map[key]
    for key in BACKBONES:
        df_out[f"stage2_conf_{key}"] = np.max(test_stage2_prob_map[key], axis=1)
        df_out[f"stage2_pred_{key}"] = [DEFECT_CLASS_LIST[i] for i in np.argmax(test_stage2_prob_map[key], axis=1)]


    save_csv(df_out, OUT_TEST_CSV)


    # ---------- 10.10 Reports ----------
    stage1_false_negative_analysis(df_out, none_class_name=NONE_CLASS_NAME)
    evaluate_stage1_binary_report(true_folder, stage1_pred_name, title="TEST Stage1 None vs Defect")


    # Stage2 defect-only report：只看 true defect 且 Stage1 沒跳過的樣本。
    df2 = df_out[
        (df_out["true_folder"].astype(str) != NONE_CLASS_NAME) &
        (df_out["stage1_pred"].astype(str) == "Defect")
    ].copy()
    if len(df2) > 0:
        print("\n==============================")
        print("TEST Stage2 Report (defect-only, after Stage1 pass)")
        print("==============================")
        print(classification_report(
            df2["true_folder"].astype(str),
            df2["stage2_pred"].astype(str),
            labels=DEFECT_CLASS_LIST,
            target_names=DEFECT_CLASS_LIST,
            digits=4,
            zero_division=0,
        ))
        cm2 = confusion_matrix(df2["true_folder"].astype(str), df2["stage2_pred"].astype(str), labels=DEFECT_CLASS_LIST)
        plot_cm(cm2, DEFECT_CLASS_LIST, "TEST Stage2 Confusion Matrix", normalize="true", show_counts=False)


    evaluate_final_report(true_folder, final_pred, title="TEST Final Two-stage Confusion Matrix")


    # ---------- 10.11 弱勢類別分析 ----------
    # 挑選目前 Stage2 單一第一名模型表現較差的幾個類別，
    # 比較 Greedy-Based Selection 選出的模型集合是否優於單一模型。
    analyze_weak_classes_against_best_single(
        df_out=df_out,
        best_single_model=best_single_stage2_model,
        selected_stage2=selected_stage2,
        top_k=WEAK_CLASS_TOP_K,
    )


    # ---------- 10.12 sample60 ----------
    df_pick = build_or_load_sample60(df_out, force_rebuild=FORCE_REBUILD_SAMPLE60)
    print("\n==============================")
    print("Fixed sample60 result")
    print("==============================")
    print(df_pick[["file", "true_folder", "final_pred", "correct_1or0"]].head(20))


    # ---------- 10.13 儲存最終設定 ----------
    final_config = {
        "selected_stage1_models": selected_stage1,
        "stage1_weights": stage1_weights,
        "stage1_threshold": thresh_stage1,
        "selected_stage2_models": selected_stage2,
        "stage2_weights": stage2_weights,
        "best_single_stage2_model": best_single_stage2_model,
        "greedy_algorithm": "sequential_f1_ratio_based_soft_voting",
        "weak_class_top_k": WEAK_CLASS_TOP_K,
        "model_specs": {
            k: {
                "display_name": v.display_name,
                "family": v.family,
                "image_size": list(v.image_size),
                "lr": v.lr,
                "stage1_weight_path": v.stage1_weight_path,
                "stage2_weight_path": v.stage2_weight_path,
                "best_stage1_pipeline": v.best_stage1_pipeline,
                "best_stage2_pipeline": v.best_stage2_pipeline,
            }
            for k, v in MODEL_SPECS.items()
        },
        "cnn_image_size": list(CNN_IMAGE_SIZE),
        "densenet_loss_weight_strength": DENSENET_LOSS_WEIGHT_STRENGTH,
        "cnn_dropout": CNN_DROPOUT,
    }
    save_json(final_config, OUT_FINAL_CONFIG_JSON)


    print("\n[INFO] CNN + DenseNet + Swin three-model ensemble experiment done.")




if __name__ == "__main__":
    main()
```
